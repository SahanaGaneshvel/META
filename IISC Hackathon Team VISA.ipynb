```python
import pandas as pd
import numpy as np
import torch
from tsfm_public import (
    TimeSeriesForecastingPipeline,
    TinyTimeMixerForPrediction,
    TimeSeriesPreprocessor,
)
pd.set_option("display.max_columns", 200)

# Determinism
seed = 42
np.random.seed(seed); torch.manual_seed(seed)
torch.set_float32_matmul_precision("high")
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
# Hour-of-Day baseline (fixed)
# Ensure timestamp is datetime
test_df = test_df.copy()
test_df["timestamp"] = pd.to_datetime(test_df["timestamp"])

# Build hour-of-day mean from the 168h inputs per window_id
inputs = (
    test_df.query("role == 'input'")
           .assign(hod=lambda x: x["timestamp"].dt.hour)
)

hod_means = (
    inputs.groupby(["window_id", "hod"])["meter_reading"]
          .mean()
          .rename("hod_mean")
)

targets = (
    test_df.query("role == 'target'")
           .assign(hod=lambda x: x["timestamp"].dt.hour)
           [["row_id", "window_id", "hod"]]
)

sub_hod = targets.merge(hod_means, on=["window_id", "hod"], how="left")

# If any hod bucket is missing, fall back to window mean
wmean = inputs.groupby("window_id")["meter_reading"].mean().rename("wmean")
sub_hod = sub_hod.merge(wmean, on="window_id", how="left")

sub_hod["meter_reading"] = (
    sub_hod["hod_mean"].fillna(sub_hod["wmean"]).clip(lower=0)
)

sub_hod = (
    sub_hod[["row_id", "meter_reading"]]
    .sort_values("row_id")
    .reset_index(drop=True)
)

hod_path = f"{OUTPUT_DIR}/submission_hod.csv"
sub_hod.to_csv(hod_path, index=False)
print("Wrote:", hod_path, "| shape:", sub_hod.shape)
```

```python
# Preprocessor config
timestamp_column = "timestamp"
target_columns = ["meter_reading"]
context_length = 180       # TTM r2 checkpoint expects 180h context
forecast_length = 60       # ...and predicts 60 steps
id_columns = ["building_id", "window_id"]  # per-series scaling

# Fit scaler on all INPUT rows (safe)
tsp = TimeSeriesPreprocessor(
    timestamp_column=timestamp_column,
    id_columns=id_columns,
    target_columns=target_columns,
    control_columns=[],
    context_length=context_length,
    forecast_length=forecast_length,
    scaling=True,
    scaler_type="standard",
    freq="h",
)
tsp.train(test_df.query("role == 'input'"))

# Load model
ttm = TinyTimeMixerForPrediction.from_pretrained(
    "ibm-granite/granite-timeseries-ttm-r2",
    revision="180-60-ft-l1-r2.1",
    num_input_channels=len(target_columns),
).eval()

pipeline = TimeSeriesForecastingPipeline(
    ttm, explode_forecasts=True, feature_extractor=tsp, device=device
)

# Helper: pad each window from 168h to 180h by repeating the first 12 hours
def pad_to_180(win_df: pd.DataFrame) -> pd.DataFrame:
    win = win_df.sort_values(timestamp_column).copy()
    if len(win) == 168:
        pad = win.iloc[:12].copy()
        pad[timestamp_column] = pad[timestamp_column] - pd.to_timedelta(
            np.arange(12, 0, -1), unit="h"
        )
        win = pd.concat([pad, win], ignore_index=True).sort_values(timestamp_column)
    return win

# Forecast per window
ttm_rows = []
for wi, wdf in test_df.groupby("window_id", sort=False):
    current_inp = wdf[wdf["role"]=="input"].copy()
    current_out = wdf[wdf["role"]=="target"][["row_id", "timestamp"]].copy()

    inp_padded = pad_to_180(current_inp)
    pred = pipeline(inp_padded)           # returns 60-step forecast
    pred24 = pred.iloc[:24][["meter_reading"]].reset_index(drop=True)  # take first 24

    tmp = current_out.reset_index(drop=True)
    tmp["meter_reading_ttm"] = pred24["meter_reading"].astype(float).clip(lower=0)
    ttm_rows.append(tmp)

sub_ttm = pd.concat(ttm_rows, ignore_index=True)
ttm_path = f"{OUTPUT_DIR}/submission_ttm.csv"
sub_ttm.rename(columns={"meter_reading_ttm":"meter_reading"})[["row_id","meter_reading"]].to_csv(ttm_path, index=False)
print("Wrote:", ttm_path, "| shape:", sub_ttm.shape)
```

```python
# Robust blend: rename columns, merge, compute final meter_reading

# 1) Ensure expected column names
if "meter_reading" in sub_hod.columns and "meter_reading_hod" not in sub_hod.columns:
    sub_hod = sub_hod.rename(columns={"meter_reading": "meter_reading_hod"})

# sub_ttm from the TTM step typically has ['row_id','timestamp','meter_reading_ttm'].
# If it was saved as 'meter_reading', rename it too:
if "meter_reading" in sub_ttm.columns and "meter_reading_ttm" not in sub_ttm.columns:
    sub_ttm = sub_ttm.rename(columns={"meter_reading": "meter_reading_ttm"})

# 2) Merge on row_id
sub_blend = sub_hod.merge(sub_ttm[["row_id", "meter_reading_ttm"]], on="row_id", how="left")

# 3) Blend
ALPHA = 0.7  # weight on TTM (tune later if needed)
sub_blend["meter_reading_hod"] = sub_blend["meter_reading_hod"].astype(float)
sub_blend["meter_reading"] = (
    ALPHA * sub_blend["meter_reading_ttm"].fillna(sub_blend["meter_reading_hod"]) +
    (1 - ALPHA) * sub_blend["meter_reading_hod"]
).clip(lower=0)

# 4) Save submission
blend_path = f"{OUTPUT_DIR}/submission_blend.csv"
sub_blend[["row_id", "meter_reading"]].sort_values("row_id").to_csv(blend_path, index=False)
print("Wrote:", blend_path, "| shape:", sub_blend.shape)
```

```python
import pandas as pd, numpy as np
from tqdm import tqdm

# Reuse the same pad function; define if not present
def pad_to_180(win_df: pd.DataFrame, ts_col="timestamp") -> pd.DataFrame:
    win = win_df.sort_values(ts_col).copy()
    if len(win) == 168:
        pad = win.iloc[:12].copy()
        pad[ts_col] = pad[ts_col] - pd.to_timedelta(np.arange(12, 0, -1), unit="h")
        win = pd.concat([pad, win], ignore_index=True).sort_values(ts_col)
    return win

# Tag each window's targets with horizon offset 0..23
def add_hoffset(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out = out.sort_values(["window_id", "timestamp"])
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

# Quick TTM forecaster for a list of window_ids on a given dataframe (train or test)
def ttm_forecast_windows(df: pd.DataFrame, window_ids, pipeline, steps=24):
    preds = []
    for wi in tqdm(window_ids, desc="TTM windows"):
        wdf = df[df["window_id"]==wi]
        inp = wdf[wdf["role"]=="input"].copy()
        tar = wdf[wdf["role"]=="target"][["timestamp"]].copy()
        pred = pipeline(pad_to_180(inp)).iloc[:steps][["meter_reading"]].reset_index(drop=True)
        tar = tar.reset_index(drop=True)
        tar["meter_reading_ttm"] = pred["meter_reading"].astype(float)
        tar["window_id"] = wi
        preds.append(tar)
    return pd.concat(preds, ignore_index=True)
```

```python
import pandas as pd, numpy as np
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error

# 0) Start from your existing 'feat' frame that already has:
#    columns: ['building_id','window_id','timestamp','h_offset','ytrue','y_hod','y_ttm'] (y_ttm may be NaN for some rows)
# If you lost it, rebuild as in earlier cells; otherwise continue.

# 1) Extra baselines on TRAIN
trin = train_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].copy()
trin["hod"] = trin["timestamp"].dt.hour

# Weekly naive
idx_in = trin.set_index(["window_id","timestamp"])["meter_reading"].rename("y_lag168")
wk = feat[["window_id","timestamp"]].copy()
wk["timestamp_lag"] = wk["timestamp"] - pd.Timedelta(hours=168)
wk = wk.join(idx_in, on=["window_id","timestamp_lag"])

# Lag-24
idx24 = idx_in.rename("mr").rename(None)
lag24 = feat[["window_id","timestamp"]].copy()
lag24["timestamp_lag"] = lag24["timestamp"] - pd.Timedelta(hours=24)
lag24 = lag24.join(idx_in, on=["window_id","timestamp_lag"]).rename(columns={"y_lag168":"y_lag24"})

# Last value in window (persistence)
last_val = (trin.sort_values("timestamp")
                 .groupby("window_id").tail(1)
                 .set_index("window_id")["meter_reading"].rename("y_last"))
feat = feat.merge(wk[["window_id","timestamp","y_lag168"]], on=["window_id","timestamp"], how="left")
feat = feat.merge(lag24[["window_id","timestamp","y_lag24"]], on=["window_id","timestamp"], how="left")
feat = feat.join(last_val, on="window_id")

# 2) Clean types & fallbacks
for c in ["ytrue","y_hod","y_ttm","y_lag168","y_lag24","y_last"]:
    feat[c] = pd.to_numeric(feat[c], errors="coerce")

# Fill NA bases with y_hod (safe shrinkage), then clip ≥0
for c in ["y_ttm","y_lag168","y_lag24","y_last"]:
    feat[c] = feat[c].fillna(feat["y_hod"]).clip(lower=0)

# 3) Learn non-negative ridge weights per (building_id, h_offset)
models = []
colsX = ["y_ttm","y_hod","y_lag168","y_lag24","y_last"]

for (bid, hoff), g in feat.dropna(subset=["ytrue","y_hod"]).groupby(["building_id","h_offset"], sort=False):
    X = g[colsX].values
    y = g["ytrue"].values
    # Ridge with positive coefficients; fit_intercept=False (we’ll add bias separately)
    reg = Ridge(alpha=1e-3, fit_intercept=False, positive=True)
    reg.fit(X, y)
    w = reg.coef_.astype(float)
    # Renormalize to sum-to-one (convex weights)
    s = w.sum()
    if s > 0:
        w = w / s
    models.append([bid, hoff] + list(w))

weight_table = pd.DataFrame(models, columns=["building_id","h_offset"] + [f"w_{c}" for c in colsX])
print("Weight table:", weight_table.shape)

# 4) Bias (mean residual) with the learned weights
tmp = feat.merge(weight_table, on=["building_id","h_offset"], how="left")
for c in colsX:
    tmp[f"w_{c}"] = tmp[f"w_{c}"].fillna(0.0)
# If weights missing, default to HoD
tmp["w_y_hod"] = np.where(tmp[[f"w_{c}" for c in colsX]].sum(axis=1)==0, 1.0, tmp["w_y_hod"])

yhat = (tmp["w_y_ttm"]*tmp["y_ttm"] +
        tmp["w_y_hod"]*tmp["y_hod"] +
        tmp["w_y_lag168"]*tmp["y_lag168"] +
        tmp["w_y_lag24"]*tmp["y_lag24"] +
        tmp["w_y_last"]*tmp["y_last"]).clip(lower=0)

resid = tmp["ytrue"] - yhat
bias_table = (pd.DataFrame({"building_id": tmp["building_id"],
                            "h_offset": tmp["h_offset"],
                            "bias": resid})
              .groupby(["building_id","h_offset"])["bias"]
              .mean().reset_index())
print("Bias table:", bias_table.shape)
```

```python
# 1) Build all bases on TEST
test_inp = test_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].copy()
test_inp["hod"] = test_inp["timestamp"].dt.hour

# HoD (already have sub_hod)
if "meter_reading" in sub_hod.columns and "meter_reading_hod" not in sub_hod.columns:
    sub_hod = sub_hod.rename(columns={"meter_reading":"meter_reading_hod"})

# TTM (already have sub_ttm)
if "meter_reading" in sub_ttm.columns and "meter_reading_ttm" not in sub_ttm.columns:
    sub_ttm = sub_ttm.rename(columns={"meter_reading":"meter_reading_ttm"})

# Weekly naive
idx_in_test = test_inp.set_index(["window_id","timestamp"])["meter_reading"].rename("y_lag168")
wk_te = test_df.query("role=='target'")[["row_id","window_id","timestamp"]].copy()
wk_te["timestamp_lag"] = wk_te["timestamp"] - pd.Timedelta(hours=168)
wk_te = wk_te.join(idx_in_test, on=["window_id","timestamp_lag"])

# Lag-24
lag24_te = test_df.query("role=='target'")[["row_id","window_id","timestamp"]].copy()
lag24_te["timestamp_lag"] = lag24_te["timestamp"] - pd.Timedelta(hours=24)
lag24_te = lag24_te.join(idx_in_test, on=["window_id","timestamp_lag"]).rename(columns={"y_lag168":"y_lag24"})

# Last value per window
last_val_te = (test_inp.sort_values("timestamp")
                    .groupby("window_id").tail(1)
                    .set_index("window_id")["meter_reading"].rename("y_last"))

# 2) Join everything + add h_offset and building_id
def add_hoffset(df):
    out = df.sort_values(["window_id","timestamp"]).copy()
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

test_targets = test_df.query("role=='target'")[["row_id","window_id","building_id","timestamp"]]
test_targets = add_hoffset(test_targets)

blend = (test_targets
         .merge(sub_hod, on="row_id", how="left")
         .merge(sub_ttm[["row_id","meter_reading_ttm"]], on="row_id", how="left")
         .merge(wk_te[["row_id","y_lag168"]], on="row_id", how="left")
         .merge(lag24_te[["row_id","y_lag24"]], on="row_id", how="left")
         .join(last_val_te, on="window_id"))

# Fill NA bases with HoD
blend["y_hod"] = pd.to_numeric(blend["meter_reading_hod"], errors="coerce")
for c_old, c_new in [("meter_reading_ttm","y_ttm"), ("y_lag168","y_lag168"), ("y_lag24","y_lag24"), ("y_last","y_last")]:
    blend[c_new] = pd.to_numeric(blend[c_old], errors="coerce")

for c in ["y_ttm","y_lag168","y_lag24","y_last"]:
    blend[c] = blend[c].fillna(blend["y_hod"]).clip(lower=0)

# 3) Attach weights + bias
blend = blend.merge(weight_table, on=["building_id","h_offset"], how="left") \
             .merge(bias_table, on=["building_id","h_offset"], how="left")

# Defaults where weights missing: pure HoD, zero bias
for c in [f"w_{k}" for k in ["y_ttm","y_hod","y_lag168","y_lag24","y_last"]]:
    blend[c] = blend[c].fillna(0.0)
row_sum = blend[[f"w_{k}" for k in ["y_ttm","y_hod","y_lag168","y_lag24","y_last"]]].sum(axis=1)
# If no weights found, set w_hod=1
blend.loc[row_sum.eq(0), "w_y_hod"] = 1.0
blend["bias"] = blend["bias"].fillna(0.0)

# 4) Final prediction (convex weights + bias), clamp ≥0
pred = (blend["w_y_ttm"]*blend["y_ttm"] +
        blend["w_y_hod"]*blend["y_hod"] +
        blend["w_y_lag168"]*blend["y_lag168"] +
        blend["w_y_lag24"]*blend["y_lag24"] +
        blend["w_y_last"]*blend["y_last"] + blend["bias"]).clip(lower=0)

sub_final = pd.DataFrame({"row_id": blend["row_id"], "meter_reading": pred}).sort_values("row_id")
out_path = f"{OUTPUT_DIR}/submission_convex5_tuned_bias.csv"
sub_final.to_csv(out_path, index=False)
print("Wrote:", out_path, "| rows:", len(sub_final))
```

```python
# Rebuild alphas & bias without redoing everything else
import pandas as pd, numpy as np, torch
from tsfm_public import TimeSeriesPreprocessor, TinyTimeMixerForPrediction, TimeSeriesForecastingPipeline
from sklearn.metrics import mean_squared_error
from tqdm import tqdm

# Helpers (define if missing)
def add_hoffset(df: pd.DataFrame) -> pd.DataFrame:
    out = df.sort_values(["window_id","timestamp"]).copy()
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

def pad_to_180(win_df: pd.DataFrame, ts_col="timestamp") -> pd.DataFrame:
    win = win_df.sort_values(ts_col).copy()
    if len(win) == 168:
        pad = win.iloc[:12].copy()
        pad[ts_col] = pad[ts_col] - pd.to_timedelta(np.arange(12, 0, -1), unit="h")
        win = pd.concat([pad, win], ignore_index=True).sort_values(ts_col)
    return win

def ttm_forecast_windows(df: pd.DataFrame, window_ids, pipeline, steps=24):
    preds = []
    for wi in tqdm(window_ids, desc="TTM windows"):
        wdf = df[df["window_id"]==wi]
        inp = wdf[wdf["role"]=="input"].copy()
        tar = wdf[wdf["role"]=="target"][["timestamp"]].copy()
        pred = pipeline(pad_to_180(inp)).iloc[:steps][["meter_reading"]].reset_index(drop=True)
        tar = tar.reset_index(drop=True)
        tar["meter_reading_ttm"] = pred["meter_reading"].astype(float)
        tar["window_id"] = wi
        preds.append(tar)
    return pd.concat(preds, ignore_index=True)

# Ensure datetimes
train_df["timestamp"] = pd.to_datetime(train_df["timestamp"])

# 1) Build HoD target frame on TRAIN
train_tgt = (train_df.query("role=='target'")[["window_id","building_id","timestamp","meter_reading"]]
             .rename(columns={"meter_reading":"ytrue"}))
train_tgt = add_hoffset(train_tgt)
train_tgt["hod"] = train_tgt["timestamp"].dt.hour

trin = (train_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].copy())
trin["hod"] = pd.to_datetime(trin["timestamp"]).dt.hour
hod_means = (trin.groupby(["window_id","hod"])["meter_reading"].mean()
             .rename("y_hod").reset_index())
wmean_tr = (trin.groupby("window_id")["meter_reading"].mean()
            .rename("wmean").reset_index())

feat = (train_tgt.merge(hod_means, on=["window_id","hod"], how="left")
                 .merge(wmean_tr, on="window_id", how="left"))
feat["y_hod"] = feat["y_hod"].fillna(feat["wmean"])

# 2) Build TRAIN TTM predictions on a sample of windows (fast and sufficient)
N_WINDOWS = 1200  # raise to ~2000 if you have time/GPU
sample_wids = train_df["window_id"].drop_duplicates().sample(N_WINDOWS, random_state=42).tolist()

tsp_train = TimeSeriesPreprocessor(
    timestamp_column="timestamp",
    id_columns=["building_id","window_id"],
    target_columns=["meter_reading"],
    control_columns=[],
    context_length=180,
    forecast_length=60,
    scaling=True,
    scaler_type="standard",
    freq="h",
)
tsp_train.train(train_df.query("role=='input' and window_id in @sample_wids"))

ttm_train = TinyTimeMixerForPrediction.from_pretrained(
    "ibm-granite/granite-timeseries-ttm-r2",
    revision="180-60-ft-l1-r2.1",
    num_input_channels=1,
).eval()

device = "cuda" if torch.cuda.is_available() else "cpu"
pipeline_tr = TimeSeriesForecastingPipeline(ttm_train, explode_forecasts=True, feature_extractor=tsp_train, device=device)

pred_ttm = ttm_forecast_windows(train_df, sample_wids, pipeline_tr, steps=24)
feat = feat.merge(pred_ttm, on=["window_id","timestamp"], how="left")
feat["y_ttm"] = pd.to_numeric(feat["meter_reading_ttm"], errors="coerce")
feat["ytrue"] = pd.to_numeric(feat["ytrue"], errors="coerce")
feat["y_hod"] = pd.to_numeric(feat["y_hod"], errors="coerce")

# 3) Learn per (building_id, h_offset) alpha
alphas = np.linspace(0, 1, 21)
best = []
for (bid, hoff), g in feat.dropna(subset=["ytrue","y_hod"]).groupby(["building_id","h_offset"], sort=False):
    yh = g["y_hod"].to_numpy()
    yt = g["y_ttm"].fillna(g["y_hod"]).to_numpy()  # fallback to HoD when TTM missing
    y  = g["ytrue"].to_numpy()
    yh = np.clip(yh, 0, None); yt = np.clip(yt, 0, None)
    best_a, best_m = None, None
    for a in alphas:
        yhat = np.clip(a*yt + (1-a)*yh, 0, None)
        m = mean_squared_error(y, yhat)
        if best_m is None or m < best_m:
            best_a, best_m = float(a), m
    best.append((bid, hoff, best_a, best_m))

alpha_table = pd.DataFrame(best, columns=["building_id","h_offset","alpha","cv_mse"])
print("alpha_table:", alpha_table.shape)

# 4) Bias per (building_id, h_offset)
feat_bias = feat.dropna(subset=["ytrue","y_hod"]).merge(alpha_table, on=["building_id","h_offset"], how="left")
feat_bias["alpha"] = feat_bias["alpha"].fillna(0.0)
yhat_train = (feat_bias["alpha"] * feat_bias["y_ttm"].fillna(feat_bias["y_hod"])
              + (1 - feat_bias["alpha"]) * feat_bias["y_hod"]).clip(lower=0)
resid = feat_bias["ytrue"] - yhat_train
bias_table = (pd.DataFrame({"building_id": feat_bias["building_id"], "h_offset": feat_bias["h_offset"], "bias": resid})
              .groupby(["building_id","h_offset"])["bias"].mean().reset_index())
print("bias_table:", bias_table.shape)

# Save for future recovery
alpha_table.to_parquet(f"{DATA_DIR}/alpha_table.parquet", index=False)
bias_table.to_parquet(f"{DATA_DIR}/bias_table.parquet", index=False)
print("Saved alpha_table.parquet and bias_table.parquet to", DATA_DIR)
```

```python
# Closed-form alpha on TRAIN: use TTM vs HoD where TTM preds exist; clip to [0,1], shrink by support,
# smooth across horizons per building, then compute a shrunk bias.

import pandas as pd, numpy as np, torch
from tsfm_public import TimeSeriesPreprocessor, TinyTimeMixerForPrediction, TimeSeriesForecastingPipeline
from tqdm import tqdm
from sklearn.metrics import mean_squared_error

# --- helpers ---
def add_hoffset(df: pd.DataFrame) -> pd.DataFrame:
    out = df.sort_values(["window_id","timestamp"]).copy()
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

def pad_to_180(win_df: pd.DataFrame, ts_col="timestamp") -> pd.DataFrame:
    win = win_df.sort_values(ts_col).copy()
    if len(win) == 168:
        pad = win.iloc[:12].copy()
        pad[ts_col] = pad[ts_col] - pd.to_timedelta(np.arange(12, 0, -1), unit="h")
        win = pd.concat([pad, win], ignore_index=True).sort_values(ts_col)
    return win

def ttm_forecast_windows(df: pd.DataFrame, window_ids, pipeline, steps=24):
    preds = []
    for wi in tqdm(window_ids, desc="TTM windows"):
        wdf = df[df["window_id"]==wi]
        inp = wdf[wdf["role"]=="input"].copy()
        tar = wdf[wdf["role"]=="target"][["timestamp"]].copy()
        pred = pipeline(pad_to_180(inp)).iloc[:steps][["meter_reading"]].reset_index(drop=True)
        tar = tar.reset_index(drop=True)
        tar["meter_reading_ttm"] = pred["meter_reading"].astype(float)
        tar["window_id"] = wi
        preds.append(tar)
    return pd.concat(preds, ignore_index=True)

# Ensure datetime
train_df["timestamp"] = pd.to_datetime(train_df["timestamp"])

# 1) TRAIN targets + HoD baseline
train_tgt = (train_df.query("role=='target'")[["window_id","building_id","timestamp","meter_reading"]]
             .rename(columns={"meter_reading":"ytrue"}))
train_tgt = add_hoffset(train_tgt)
train_tgt["hod"] = train_tgt["timestamp"].dt.hour

trin = train_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].copy()
trin["hod"] = pd.to_datetime(trin["timestamp"]).dt.hour
y_hod_tab = (trin.groupby(["window_id","hod"])["meter_reading"].mean().rename("y_hod").reset_index())
wmean_tr  = (trin.groupby("window_id")["meter_reading"].mean().rename("wmean").reset_index())

feat = (train_tgt.merge(y_hod_tab, on=["window_id","hod"], how="left")
                 .merge(wmean_tr,  on="window_id",      how="left"))
feat["y_hod"] = feat["y_hod"].fillna(feat["wmean"]).astype(float)
feat["ytrue"] = feat["ytrue"].astype(float)

# 2) TTM preds on a moderate sample of windows (enough for stable α)
N_WINDOWS = 1500   # ~1–3 minutes on CPU; adjust if needed
sample_wids = train_df["window_id"].drop_duplicates().sample(N_WINDOWS, random_state=42).tolist()

tsp_train = TimeSeriesPreprocessor(
    timestamp_column="timestamp",
    id_columns=["building_id","window_id"],
    target_columns=["meter_reading"],
    control_columns=[],
    context_length=180,
    forecast_length=60,
    scaling=True,
    scaler_type="standard",
    freq="h",
)
tsp_train.train(train_df.query("role=='input' and window_id in @sample_wids"))

ttm_train = TinyTimeMixerForPrediction.from_pretrained(
    "ibm-granite/granite-timeseries-ttm-r2",
    revision="180-60-ft-l1-r2.1",
    num_input_channels=1,
).eval()

device = "cuda" if torch.cuda.is_available() else "cpu"
pipeline_tr = TimeSeriesForecastingPipeline(ttm_train, explode_forecasts=True, feature_extractor=tsp_train, device=device)

pred_ttm = ttm_forecast_windows(train_df, sample_wids, pipeline_tr, steps=24)
feat = feat.merge(pred_ttm, on=["window_id","timestamp"], how="left")
feat["y_ttm"] = pd.to_numeric(feat["meter_reading_ttm"], errors="coerce")

# 3) Closed-form α per (building_id, h_offset):
#    Let d = y_ttm - y_hod ; yhat = y_hod + a*d ; minimize MSE => a* = sum((y - y_hod)*d)/sum(d^2)
#    Guardrails: use only rows with y_ttm; clip a to [0,1]; shrink by support n: a = a_clipped * n/(n+L)
L_ALPHA = 48      # shrinkage strength for alpha (tune 24–96)
MIN_N   = 20      # minimum rows to trust alpha
eps     = 1e-9

rows = []
for (bid, hoff), g in feat.dropna(subset=["ytrue","y_hod","y_ttm"]).groupby(["building_id","h_offset"], sort=False):
    y  = g["ytrue"].to_numpy()
    yh = g["y_hod"].to_numpy()
    yt = g["y_ttm"].to_numpy()
    d  = yt - yh
    num = np.sum((y - yh) * d)
    den = np.sum(d * d) + eps
    a_raw = num / den
    a_clip = float(np.clip(a_raw, 0.0, 1.0))
    n = len(g)
    a_shr = a_clip * (n / (n + L_ALPHA)) if n >= MIN_N else 0.0  # fall back to HoD if too few samples
    rows.append((bid, hoff, n, a_raw, a_clip, a_shr))

alpha_table = pd.DataFrame(rows, columns=["building_id","h_offset","n","alpha_raw","alpha_clip","alpha"]).drop(columns=["alpha_raw","alpha_clip"])
print("alpha_table:", alpha_table.shape)

# 4) Smooth α across horizons per building (3-point moving average)
alpha_table = alpha_table.sort_values(["building_id","h_offset"]).copy()
alpha_table["alpha_sm"] = (
    alpha_table.groupby("building_id")["alpha"]
               .apply(lambda s: pd.Series(np.convolve(s.to_numpy(), np.array([0.2,0.6,0.2]), mode="same"), index=s.index))
               .values
)
alpha_table["alpha"] = alpha_table["alpha_sm"].clip(0,1)
alpha_table = alpha_table.drop(columns=["alpha_sm","n"])  # keep compact

# 5) Bias with shrinkage per (building_id, h_offset)
tmp = feat.merge(alpha_table, on=["building_id","h_offset"], how="left")
tmp["alpha"] = tmp["alpha"].fillna(0.0)
yhat = (tmp["alpha"] * tmp["y_ttm"].fillna(tmp["y_hod"]) + (1-tmp["alpha"]) * tmp["y_hod"]).clip(lower=0)
resid = tmp["ytrue"] - yhat
bias = pd.DataFrame({"building_id": tmp["building_id"], "h_offset": tmp["h_offset"], "resid": resid})
bias_table = bias.groupby(["building_id","h_offset"])["resid"].agg(["mean","count"]).reset_index()
L_BIAS = 50  # shrinkage for bias
bias_table["bias"] = bias_table["mean"] * (bias_table["count"] / (bias_table["count"] + L_BIAS))
bias_table = bias_table[["building_id","h_offset","bias"]]

# Save for reuse
alpha_table.to_parquet(f"{DATA_DIR}/alpha_table_cf.parquet", index=False)
bias_table.to_parquet(f"{DATA_DIR}/bias_table_cf.parquet", index=False)
print("Saved alpha_table_cf.parquet & bias_table_cf.parquet")
```

```python
# Learn a small residual corrector: y_final = y_blend + beta * (lag24 - y_hod)
# Uses your existing alpha_table_cf + bias_table_cf (closed-form α + shrunk bias).

import pandas as pd, numpy as np

DATA_DIR = "/content/drive/My Drive/IISC_data"

# Helpers
def add_hoffset(df: pd.DataFrame) -> pd.DataFrame:
    out = df.sort_values(["window_id","timestamp"]).copy()
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

# Make sure frames are in memory
if "train_df" not in globals():
    train_df = pd.read_csv(f"{DATA_DIR}/train.csv", parse_dates=["timestamp"])
if "alpha_table" not in globals():
    alpha_table = pd.read_parquet(f"{DATA_DIR}/alpha_table_cf.parquet")
if "bias_table" not in globals():
    bias_table  = pd.read_parquet(f"{DATA_DIR}/bias_table_cf.parquet")

# --- Build TRAIN target rows with HoD, lag24, TTM, blend ---
trin = train_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].copy()
trin["hod"] = trin["timestamp"].dt.hour

# HoD mean within each window for each hour
y_hod_tab = (trin.groupby(["window_id","hod"])["meter_reading"]
             .mean().rename("y_hod").reset_index())

# TRAIN targets + horizon index
tgt = (train_df.query("role=='target'")[["window_id","building_id","timestamp","meter_reading"]]
       .rename(columns={"meter_reading":"ytrue"}))
tgt = add_hoffset(tgt)
tgt["hod"] = tgt["timestamp"].dt.hour

feat = tgt.merge(y_hod_tab, on=["window_id","hod"], how="left")
# fallback to window mean if some hod missing (rare)
wmean_tr = trin.groupby("window_id")["meter_reading"].mean().rename("wmean").reset_index()
feat = feat.merge(wmean_tr, on="window_id", how="left")
feat["y_hod"] = feat["y_hod"].fillna(feat["wmean"])

# lag-24 from inputs
idx_in = trin.set_index(["window_id","timestamp"])["meter_reading"].rename("lag24_val")
lag24 = feat[["window_id","timestamp"]].copy()
lag24["timestamp_lag"] = lag24["timestamp"] - pd.Timedelta(hours=24)
lag24 = lag24.join(idx_in, on=["window_id","timestamp_lag"])
feat = feat.merge(lag24[["window_id","timestamp","lag24_val"]], on=["window_id","timestamp"], how="left")

# bring in TTM (use the same trick: closed-form α blend uses TTM only to set α;
# we don't need to recompute TTM here—just rebuild y_blend from α+bias and HoD/TTM from TRAIN)
# For stability, build a minimal TTM on a sample exactly as in your α step:
from tsfm_public import TimeSeriesPreprocessor, TinyTimeMixerForPrediction, TimeSeriesForecastingPipeline
import torch
from tqdm import tqdm

def pad_to_180(win_df: pd.DataFrame, ts_col="timestamp") -> pd.DataFrame:
    win = win_df.sort_values(ts_col).copy()
    if len(win) == 168:
        pad = win.iloc[:12].copy()
        pad[ts_col] = pad[ts_col] - pd.to_timedelta(np.arange(12, 0, -1), unit="h")
        win = pd.concat([pad, win], ignore_index=True).sort_values(ts_col)
    return win

def ttm_forecast_windows(df: pd.DataFrame, window_ids, pipeline, steps=24):
    preds = []
    for wi in tqdm(window_ids, desc="TTM windows (β tune)"):
        wdf = df[df["window_id"]==wi]
        inp = wdf[wdf["role"]=="input"].copy()
        tar = wdf[wdf["role"]=="target"][["timestamp"]].copy()
        pred = pipeline(pad_to_180(inp)).iloc[:steps][["meter_reading"]].reset_index(drop=True)
        tar = tar.reset_index(drop=True)
        tar["meter_reading_ttm"] = pred["meter_reading"].astype(float)
        tar["window_id"] = wi
        preds.append(tar)
    return pd.concat(preds, ignore_index=True)

# Sample some windows for TTM to keep it fast (same spirit as α)
N_WINDOWS = 1500
sample_wids = train_df["window_id"].drop_duplicates().sample(N_WINDOWS, random_state=42).tolist()

tsp_tr = TimeSeriesPreprocessor(
    timestamp_column="timestamp",
    id_columns=["building_id","window_id"],
    target_columns=["meter_reading"],
    control_columns=[],
    context_length=180,
    forecast_length=60,
    scaling=True,
    scaler_type="standard",
    freq="h",
)
tsp_tr.train(train_df.query("role=='input' and window_id in @sample_wids"))

ttm_tr = TinyTimeMixerForPrediction.from_pretrained(
    "ibm-granite/granite-timeseries-ttm-r2",
    revision="180-60-ft-l1-r2.1",
    num_input_channels=1,
).eval()

device = "cuda" if torch.cuda.is_available() else "cpu"
pipe_tr = TimeSeriesForecastingPipeline(ttm_tr, explode_forecasts=True, feature_extractor=tsp_tr, device=device)

pred_ttm = ttm_forecast_windows(train_df, sample_wids, pipe_tr, steps=24)
feat = feat.merge(pred_ttm, on=["window_id","timestamp"], how="left").rename(columns={"meter_reading_ttm":"y_ttm"})

# Compute blend on TRAIN: y_blend = α*y_ttm + (1-α)*y_hod + bias
tmp = feat.merge(alpha_table, on=["building_id","h_offset"], how="left").merge(bias_table, on=["building_id","h_offset"], how="left")
tmp["alpha"] = tmp["alpha"].fillna(0.0); tmp["bias"] = tmp["bias"].fillna(0.0)
tmp["y_ttm"] = pd.to_numeric(tmp["y_ttm"], errors="coerce").fillna(tmp["y_hod"])
tmp["y_blend"] = (tmp["alpha"]*tmp["y_ttm"] + (1-tmp["alpha"])*tmp["y_hod"] + tmp["bias"]).clip(lower=0)

# Residual r, feature z = (lag24 - y_hod)
tmp["z"] = pd.to_numeric(tmp["lag24_val"], errors="coerce").fillna(tmp["y_hod"]) - tmp["y_hod"]
tmp["r"] = pd.to_numeric(tmp["ytrue"], errors="coerce") - tmp["y_blend"]

# Closed-form β per (building_id, h_offset), with shrinkage + smoothing
L_BETA = 64   # shrinkage strength (try 48–96)
MIN_N  = 20
eps    = 1e-9

rows = []
for (bid, hoff), g in tmp.dropna(subset=["r","z"]).groupby(["building_id","h_offset"], sort=False):
    z = g["z"].to_numpy()
    r = g["r"].to_numpy()
    num = np.sum(r*z); den = np.sum(z*z) + eps
    b_raw = num/den
    n = len(g)
    # clip β to avoid wild corrections; shrink toward 0
    b_clip = float(np.clip(b_raw, -0.5, 0.5))
    b_shr  = b_clip * (n / (n + L_BETA)) if n >= MIN_N else 0.0
    rows.append((bid, hoff, n, b_shr))

beta_table = pd.DataFrame(rows, columns=["building_id","h_offset","n","beta"])
beta_table = beta_table.sort_values(["building_id","h_offset"]).copy()

# Smooth β across horizons per building (gentle 3-point)
kernel = np.array([0.2, 0.6, 0.2])
beta_table["beta_sm"] = (
    beta_table.groupby("building_id")["beta"]
              .apply(lambda s: pd.Series(np.convolve(s.to_numpy(), kernel, mode="same"), index=s.index))
              .values
)
beta_table["beta"] = beta_table["beta_sm"]
beta_table = beta_table.drop(columns=["beta_sm","n"])

beta_table.to_parquet(f"{DATA_DIR}/beta_table_cf.parquet", index=False)
print("beta_table:", beta_table.shape, "saved to beta_table_cf.parquet")
```

```python
# Patch: robust window-level features + finish γ and λ, then save tables.

import numpy as np
import pandas as pd

# --- Robust window-level features from TRAIN inputs (trin) ---
w_inputs = (
    trin.sort_values(["window_id","timestamp"])
        .groupby("window_id")
        .agg(
            head24_mean=("meter_reading", lambda s: s.head(24).mean()),
            last24_mean=("meter_reading", lambda s: s.tail(24).mean()),
            x_last=("meter_reading", "last"),
        )
        .reset_index()
)
w_inputs["s_win"] = w_inputs["last24_mean"] / w_inputs["head24_mean"].clip(lower=1e-6)
w_inputs = w_inputs[["window_id","s_win","x_last"]]

# First-hour prediction y0 per window (from y_blend where h_offset==0)
y0 = (tmp.loc[tmp["h_offset"]==0, ["window_id","y_blend"]]
        .rename(columns={"y_blend":"y0"}))

win_feat = w_inputs.merge(y0, on="window_id", how="left")
win_feat["delta_win"] = (win_feat["x_last"] - win_feat["y0"]).fillna(0.0)

# Attach features to all target rows
tmp = (tmp.drop(columns=[c for c in ["s_win","delta_win"] if c in tmp.columns])
          .merge(win_feat[["window_id","s_win","delta_win"]], on="window_id", how="left"))
tmp["s_win"] = tmp["s_win"].fillna(1.0)
tmp["delta_win"] = tmp["delta_win"].fillna(0.0)

# Residual and design features
tmp["r"] = tmp["ytrue"] - tmp["y_blend"]
tmp["w_h"] = np.exp(-tmp["h_offset"] / 12.0)  # decay across horizons
tmp["phi"] = tmp["delta_win"] * tmp["w_h"]
tmp["psi"] = (tmp["s_win"] - 1.0) * tmp["y_blend"]

# --- Closed-form per-building γ (offset) and λ (scale) with shrinkage ---
L_GAMMA = 100
L_LAMBDA = 100
MIN_N = 50
eps = 1e-9

rows_g, rows_l = [], []
for bid, g in tmp.dropna(subset=["r","phi","psi"]).groupby("building_id", sort=False):
    # γ (offset)
    num_g = float(np.sum(g["r"] * g["phi"]))
    den_g = float(np.sum(g["phi"] * g["phi"])) + eps
    gamma_raw = num_g / den_g
    n_g = len(g)
    gamma_clip = float(np.clip(gamma_raw, -0.5, 0.5))
    gamma = gamma_clip * (n_g / (n_g + L_GAMMA)) if n_g >= MIN_N else 0.0
    rows_g.append((bid, gamma, n_g))

    # λ (scale): regress r / y_blend on (s_win - 1)
    yb = np.clip(g["y_blend"].to_numpy(), 1e-6, None)
    rr = (g["r"].to_numpy() / yb)
    ss = (g["s_win"].to_numpy() - 1.0)
    num_l = float(np.sum(rr * ss))
    den_l = float(np.sum(ss * ss)) + eps
    lam_raw = num_l / den_l
    n_l = len(g)
    lam_clip = float(np.clip(lam_raw, -0.5, 0.5))
    lam = lam_clip * (n_l / (n_l + L_LAMBDA)) if n_l >= MIN_N else 0.0
    rows_l.append((bid, lam, n_l))

gamma_table = pd.DataFrame(rows_g, columns=["building_id","gamma","n"])
lambda_table = pd.DataFrame(rows_l, columns=["building_id","lambda","n"])

gamma_table.to_parquet(f"{DATA_DIR}/gamma_table.parquet", index=False)
lambda_table.to_parquet(f"{DATA_DIR}/lambda_table.parquet", index=False)
print("γ learned for", len(gamma_table), "buildings; λ learned for", len(lambda_table), "buildings")
```

```python
# Learn per-(building, horizon) betas for z24=(lag24 - y_hod) and z168=(lag168 - y_hod)
# Baseline = closed-form α + shrunk bias blend (y_blend). We DO NOT include γ/λ here.

import pandas as pd, numpy as np, torch
from tsfm_public import TimeSeriesPreprocessor, TinyTimeMixerForPrediction, TimeSeriesForecastingPipeline
from tqdm import tqdm

DATA_DIR = "/content/drive/My Drive/IISC_data"

# ---------- helpers ----------
def add_hoffset(df: pd.DataFrame) -> pd.DataFrame:
    out = df.sort_values(["window_id","timestamp"]).copy()
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

def pad_to_180(win_df: pd.DataFrame, ts_col="timestamp") -> pd.DataFrame:
    win = win_df.sort_values(ts_col).copy()
    if len(win) == 168:
        pad = win.iloc[:12].copy()
        pad[ts_col] = pad[ts_col] - pd.to_timedelta(np.arange(12, 0, -1), unit="h")
        win = pd.concat([pad, win], ignore_index=True).sort_values(ts_col)
    return win

def ttm_forecast_windows(df: pd.DataFrame, window_ids, pipeline, steps=24):
    preds = []
    for wi in tqdm(window_ids, desc="TTM windows (beta tune)"):
        wdf = df[df["window_id"]==wi]
        inp = wdf[wdf["role"]=="input"].copy()
        tar = wdf[wdf["role"]=="target"][["timestamp"]].copy()
        pred = pipeline(pad_to_180(inp)).iloc[:steps][["meter_reading"]].reset_index(drop=True)
        tar = tar.reset_index(drop=True)
        tar["meter_reading_ttm"] = pred["meter_reading"].astype(float)
        tar["window_id"] = wi
        preds.append(tar)
    return pd.concat(preds, ignore_index=True)

# ---------- load data/artifacts ----------
train_df = pd.read_csv(f"{DATA_DIR}/train.csv", parse_dates=["timestamp"])
alpha_table = pd.read_parquet(f"{DATA_DIR}/alpha_table_cf.parquet")
bias_table  = pd.read_parquet(f"{DATA_DIR}/bias_table_cf.parquet")

# TRAIN targets + HoD
trin = train_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].copy()
trin["hod"] = trin["timestamp"].dt.hour
y_hod_tab = (trin.groupby(["window_id","hod"])["meter_reading"].mean()
             .rename("y_hod").reset_index())
wmean_tr = (trin.groupby("window_id")["meter_reading"].mean()
            .rename("wmean").reset_index())

tgt = (train_df.query("role=='target'")[["window_id","building_id","timestamp","meter_reading"]]
       .rename(columns={"meter_reading":"ytrue"}))
tgt = add_hoffset(tgt); tgt["hod"] = tgt["timestamp"].dt.hour

feat = (tgt.merge(y_hod_tab, on=["window_id","hod"], how="left")
           .merge(wmean_tr,  on="window_id",      how="left"))
feat["y_hod"] = feat["y_hod"].fillna(feat["wmean"]).astype(float)
feat["ytrue"] = feat["ytrue"].astype(float)

# Minimal TTM on a moderate TRAIN sample to form y_blend
N_WINDOWS = 1500
sample_wids = train_df["window_id"].drop_duplicates().sample(N_WINDOWS, random_state=42).tolist()

tsp_tr = TimeSeriesPreprocessor(
    timestamp_column="timestamp",
    id_columns=["building_id","window_id"],
    target_columns=["meter_reading"],
    control_columns=[],
    context_length=180, forecast_length=60,
    scaling=True, scaler_type="standard",
    freq="h",
)
tsp_tr.train(train_df.query("role=='input' and window_id in @sample_wids"))

ttm_tr = TinyTimeMixerForPrediction.from_pretrained(
    "ibm-granite/granite-timeseries-ttm-r2",
    revision="180-60-ft-l1-r2.1",
    num_input_channels=1,
).eval()

device = "cuda" if torch.cuda.is_available() else "cpu"
pipe_tr = TimeSeriesForecastingPipeline(ttm_tr, explode_forecasts=True, feature_extractor=tsp_tr, device=device)

pred_ttm = ttm_forecast_windows(train_df, sample_wids, pipe_tr, steps=24)  # ['window_id','timestamp','meter_reading_ttm']
feat = feat.merge(pred_ttm, on=["window_id","timestamp"], how="left").rename(columns={"meter_reading_ttm":"y_ttm"})

# y_blend (alpha + bias), no γ/λ or previous residuals
tmp = feat.merge(alpha_table, on=["building_id","h_offset"], how="left").merge(bias_table, on=["building_id","h_offset"], how="left")
tmp["alpha"] = tmp["alpha"].fillna(0.0); tmp["bias"] = tmp["bias"].fillna(0.0)
tmp["y_ttm"] = pd.to_numeric(tmp["y_ttm"], errors="coerce").fillna(tmp["y_hod"])
tmp["y_blend"] = (tmp["alpha"]*tmp["y_ttm"] + (1-tmp["alpha"])*tmp["y_hod"] + tmp["bias"]).clip(lower=0)

# Build z24 and z168 from TRAIN inputs
idx_in = trin.set_index(["window_id","timestamp"])["meter_reading"].rename("mr_in")

lag24 = tmp[["window_id","timestamp"]].copy()
lag24["timestamp_lag"] = lag24["timestamp"] - pd.Timedelta(hours=24)
lag24 = lag24.join(idx_in, on=["window_id","timestamp_lag"]).rename(columns={"mr_in":"lag24_val"})

lag168 = tmp[["window_id","timestamp"]].copy()
lag168["timestamp_lag"] = lag168["timestamp"] - pd.Timedelta(hours=168)
lag168 = lag168.join(idx_in, on=["window_id","timestamp_lag"]).rename(columns={"mr_in":"lag168_val"})

tmp = (tmp.merge(lag24[["window_id","timestamp","lag24_val"]], on=["window_id","timestamp"], how="left")
          .merge(lag168[["window_id","timestamp","lag168_val"]], on=["window_id","timestamp"], how="left"))

tmp["z24"]  = pd.to_numeric(tmp["lag24_val"],  errors="coerce").fillna(tmp["y_hod"]) - tmp["y_hod"]
tmp["z168"] = pd.to_numeric(tmp["lag168_val"], errors="coerce").fillna(tmp["y_hod"]) - tmp["y_hod"]
tmp["r"]    = tmp["ytrue"] - tmp["y_blend"]

# Closed-form betas with shrinkage + smoothing
L_BETA24  = 64
L_BETA168 = 64
MIN_N = 20
eps = 1e-9

def fit_beta(group, z_col, L):
    z = group[z_col].to_numpy()
    r = group["r"].to_numpy()
    num = float((r*z).sum())
    den = float((z*z).sum()) + eps
    b_raw = num/den
    n = len(group)
    b_clip = float(np.clip(b_raw, -0.5, 0.5))
    return b_clip * (n / (n + L)) if n >= MIN_N else 0.0, n

rows24, rows168 = [], []
for (bid, hoff), g in tmp.dropna(subset=["r","z24"]).groupby(["building_id","h_offset"], sort=False):
    b, n = fit_beta(g, "z24", L_BETA24)
    rows24.append((bid, hoff, b, n))
for (bid, hoff), g in tmp.dropna(subset=["r","z168"]).groupby(["building_id","h_offset"], sort=False):
    b, n = fit_beta(g, "z168", L_BETA168)
    rows168.append((bid, hoff, b, n))

beta24 = pd.DataFrame(rows24,  columns=["building_id","h_offset","beta24","n"])
beta168= pd.DataFrame(rows168, columns=["building_id","h_offset","beta168","n"])

# Smooth across horizons (3-pt kernel)
ker = np.array([0.2, 0.6, 0.2])
for df, col in [(beta24,"beta24"), (beta168,"beta168")]:
    df.sort_values(["building_id","h_offset"], inplace=True)
    df[col] = (df.groupby("building_id")[col]
                 .apply(lambda s: pd.Series(np.convolve(s.to_numpy(), ker, mode="same"), index=s.index))
                 .values)

beta24 = beta24[["building_id","h_offset","beta24"]]
beta168= beta168[["building_id","h_offset","beta168"]]

beta24.to_parquet(f"{DATA_DIR}/beta24_table.parquet", index=False)
beta168.to_parquet(f"{DATA_DIR}/beta168_table.parquet", index=False)
print("beta24:", beta24.shape, " beta168:", beta168.shape, " (saved)")
```

```python
# TRAIN: HoD, TTM (sample), closed-form alpha with shrink + smoothing; shrunk bias
trin = train_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].copy()
trin["hod"] = trin["timestamp"].dt.hour
yhod_tr = (trin.groupby(["window_id","hod"])["meter_reading"].mean().rename("y_hod").reset_index())
wmean_tr = (trin.groupby("window_id")["meter_reading"].mean().rename("wmean").reset_index())

tgt_tr = (train_df.query("role=='target'")[["window_id","building_id","timestamp","meter_reading"]]
          .rename(columns={"meter_reading":"ytrue"}))
tgt_tr = add_hoffset(tgt_tr); tgt_tr["hod"] = tgt_tr["timestamp"].dt.hour

feat = (tgt_tr.merge(yhod_tr, on=["window_id","hod"], how="left")
              .merge(wmean_tr, on="window_id", how="left"))
feat["y_hod"] = feat["y_hod"].fillna(feat["wmean"]).astype(float)
feat["ytrue"] = feat["ytrue"].astype(float)

# Sample windows for TTM speed (adjust N_WINDOWS if needed)
N_WINDOWS = 1500
sample_wids = train_df["window_id"].drop_duplicates().sample(N_WINDOWS, random_state=SEED).tolist()

tsp_tr = TimeSeriesPreprocessor(
    timestamp_column="timestamp",
    id_columns=["building_id","window_id"],
    target_columns=["meter_reading"],
    control_columns=[],
    context_length=180, forecast_length=60,
    scaling=True, scaler_type="standard", freq="h",
)
tsp_tr.train(train_df.query("role=='input' and window_id in @sample_wids"))

ttm_tr = TinyTimeMixerForPrediction.from_pretrained(
    "ibm-granite/granite-timeseries-ttm-r2",
    revision="180-60-ft-l1-r2.1",
    num_input_channels=1,
).eval()

pipeline_tr = TimeSeriesForecastingPipeline(ttm_tr, explode_forecasts=True, feature_extractor=tsp_tr, device=device)
pred_ttm_tr = ttm_forecast_windows(train_df, sample_wids, pipeline_tr, steps=24)
feat = feat.merge(pred_ttm_tr, on=["window_id","timestamp"], how="left").rename(columns={"meter_reading_ttm":"y_ttm"})
feat["y_ttm"] = pd.to_numeric(feat["y_ttm"], errors="coerce")

# Closed-form alpha: a* = sum((y - yh)*d)/sum(d^2); clip [0,1]; shrink; smooth
L_ALPHA, MIN_N, eps = 64, 20, 1e-9
rows = []
for (bid, hoff), g in feat.dropna(subset=["ytrue","y_hod","y_ttm"]).groupby(["building_id","h_offset"], sort=False):
    y, yh, yt = g["ytrue"].to_numpy(), g["y_hod"].to_numpy(), g["y_ttm"].to_numpy()
    d = yt - yh
    a_raw = float(((y - yh)*d).sum() / (d*d).sum() if (d*d).sum()>0 else 0.0)
    a_clip = float(np.clip(a_raw, 0.0, 1.0))
    n = len(g)
    a_shr = a_clip * (n/(n+L_ALPHA)) if n>=MIN_N else 0.0
    rows.append((bid, hoff, a_shr, n))
alpha_table = pd.DataFrame(rows, columns=["building_id","h_offset","alpha","n"]).drop(columns=["n"])

# Smooth alpha over horizons (3-pt kernel)
ker = np.array([0.2, 0.6, 0.2])
alpha_table = alpha_table.sort_values(["building_id","h_offset"]).copy()
alpha_table["alpha"] = (alpha_table.groupby("building_id")["alpha"]
                                      .apply(lambda s: pd.Series(np.convolve(s.to_numpy(), ker, mode="same"), index=s.index))
                                      .values).clip(0,1)

# Bias (shrunk mean residual)
tmp = feat.merge(alpha_table, on=["building_id","h_offset"], how="left")
tmp["alpha"] = tmp["alpha"].fillna(0.0)
yhat = (tmp["alpha"] * tmp["y_ttm"].fillna(tmp["y_hod"]) + (1-tmp["alpha"]) * tmp["y_hod"]).clip(lower=0)
resid = tmp["ytrue"] - yhat
bias_table = (pd.DataFrame({"building_id": tmp["building_id"], "h_offset": tmp["h_offset"], "resid": resid})
              .groupby(["building_id","h_offset"])["resid"]
              .agg(["mean","count"]).reset_index())
L_BIAS = 50
bias_table["bias"] = bias_table["mean"] * (bias_table["count"]/(bias_table["count"]+L_BIAS))
bias_table = bias_table[["building_id","h_offset","bias"]]

alpha_table.to_parquet(f"{DATA_DIR}/alpha_table_cf.parquet", index=False)
bias_table.to_parquet(f"{DATA_DIR}/bias_table_cf.parquet", index=False)
print("Saved:", "alpha_table_cf.parquet", "bias_table_cf.parquet")
```

```python
# Reuse 'feat' (has y_hod, ytrue, y_ttm) to build y_blend, then fit betas on z24/z168
tmp = feat.merge(alpha_table, on=["building_id","h_offset"], how="left").merge(bias_table, on=["building_id","h_offset"], how="left")
tmp["alpha"] = tmp["alpha"].fillna(0.0); tmp["bias"] = tmp["bias"].fillna(0.0)
tmp["y_ttm"] = pd.to_numeric(tmp["y_ttm"], errors="coerce").fillna(tmp["y_hod"])
tmp["y_blend"] = (tmp["alpha"]*tmp["y_ttm"] + (1-tmp["alpha"])*tmp["y_hod"] + tmp["bias"]).clip(lower=0)

# lag24 / lag168 from TRAIN inputs
idx_in = trin.set_index(["window_id","timestamp"])["meter_reading"].rename("mr_in")
lag24 = tmp[["window_id","timestamp"]].copy()
lag24["timestamp_lag"] = lag24["timestamp"] - pd.Timedelta(hours=24)
lag24 = lag24.join(idx_in, on=["window_id","timestamp_lag"]).rename(columns={"mr_in":"lag24_val"})

lag168 = tmp[["window_id","timestamp"]].copy()
lag168["timestamp_lag"] = lag168["timestamp"] - pd.Timedelta(hours=168)
lag168 = lag168.join(idx_in, on=["window_id","timestamp_lag"]).rename(columns={"mr_in":"lag168_val"})

tmp = (tmp.merge(lag24[["window_id","timestamp","lag24_val"]], on=["window_id","timestamp"], how="left")
         .merge(lag168[["window_id","timestamp","lag168_val"]], on=["window_id","timestamp"], how="left"))

# HoD baseline from inputs for residuals
yhod_inp = (trin.groupby(["window_id","hod"])["meter_reading"].mean().rename("y_hod_inp").reset_index())
tmp = tmp.merge(yhod_inp, on=["window_id","hod"], how="left").merge(wmean_tr.rename(columns={"wmean":"wmean_inp"}), on="window_id", how="left")
tmp["y_hod_inp"] = tmp["y_hod_inp"].fillna(tmp["wmean_inp"])

tmp["z24"]  = pd.to_numeric(tmp["lag24_val"],  errors="coerce").fillna(tmp["y_hod_inp"]) - tmp["y_hod_inp"]
tmp["z168"] = pd.to_numeric(tmp["lag168_val"], errors="coerce").fillna(tmp["y_hod_inp"]) - tmp["y_hod_inp"]
tmp["r"]    = tmp["ytrue"] - tmp["y_blend"]

L_BETA24, L_BETA168, MIN_N, eps = 64, 64, 20, 1e-9
def fit_beta(group, zcol, L):
    z = group[zcol].to_numpy(); r = group["r"].to_numpy()
    den = float((z*z).sum()) + eps
    b_raw = float((r*z).sum()/den)
    n = len(group)
    b_clip = float(np.clip(b_raw, -0.5, 0.5))
    return (b_clip * (n/(n+L))) if n>=MIN_N else 0.0

rows24, rows168 = [], []
for (bid, hoff), g in tmp.dropna(subset=["r","z24"]).groupby(["building_id","h_offset"], sort=False):
    rows24.append((bid, hoff, fit_beta(g,"z24", L_BETA24)))
for (bid, hoff), g in tmp.dropna(subset=["r","z168"]).groupby(["building_id","h_offset"], sort=False):
    rows168.append((bid, hoff, fit_beta(g,"z168", L_BETA168)))

beta24  = pd.DataFrame(rows24,  columns=["building_id","h_offset","beta24"])
beta168 = pd.DataFrame(rows168, columns=["building_id","h_offset","beta168"])

# Smooth across horizons
ker = np.array([0.2, 0.6, 0.2])
for df, col in [(beta24,"beta24"), (beta168,"beta168")]:
    df.sort_values(["building_id","h_offset"], inplace=True)
    df[col] = (df.groupby("building_id")[col]
                 .apply(lambda s: pd.Series(np.convolve(s.to_numpy(), ker, mode="same"), index=s.index))
                 .values)

beta24.to_parquet(f"{DATA_DIR}/beta24_table.parquet", index=False)
beta168.to_parquet(f"{DATA_DIR}/beta168_table.parquet", index=False)
print("Saved: beta24_table.parquet, beta168_table.parquet")
```

```python
# Build window features from TRAIN inputs
win_inputs = (trin.sort_values(["window_id","timestamp"])
                .groupby("window_id")
                .agg(
                    head24_mean=("meter_reading", lambda s: s.head(24).mean()),
                    last24_mean=("meter_reading", lambda s: s.tail(24).mean()),
                    x_last=("meter_reading", "last"),
                ).reset_index())
win_inputs["s_win"] = win_inputs["last24_mean"] / win_inputs["head24_mean"].clip(lower=1e-6)

# y0 (first-horizon of y_blend) per window
tmp2 = tmp.copy()
y0 = (tmp2[tmp2["h_offset"]==0][["window_id","y_blend"]].rename(columns={"y_blend":"y0"}))
win_feat_tr = win_inputs.merge(y0, on="window_id", how="left")
win_feat_tr["delta_win"] = (win_feat_tr["x_last"] - win_feat_tr["y0"]).fillna(0.0)

# Attach features back to tmp2
tmp2 = tmp2.merge(win_feat_tr[["window_id","s_win","delta_win"]], on="window_id", how="left")
tmp2["r"] = tmp2["ytrue"] - tmp2["y_blend"]

# γ and λ per building (closed-form with shrinkage)
L_GAMMA, L_LAMBDA, MIN_N, eps = 100, 100, 50, 1e-9
rows_g, rows_l = [], []
for bid, g in tmp2.dropna(subset=["r","s_win","delta_win","y_blend"]).groupby("building_id", sort=False):
    # γ: regress r on phi = delta_win * decay(h)
    w_h = np.exp(-g["h_offset"].to_numpy()/12.0)
    phi = g["delta_win"].to_numpy() * w_h
    num_g = float((g["r"].to_numpy()*phi).sum()); den_g = float((phi*phi).sum()) + eps
    gamma_raw = num_g/den_g; n_g = len(g)
    gamma = float(np.clip(gamma_raw, -0.5, 0.5)) * (n_g/(n_g+L_GAMMA)) if n_g>=MIN_N else 0.0
    rows_g.append((bid, gamma))

    # λ: regress r/yb on (s_win - 1)
    yb = np.clip(g["y_blend"].to_numpy(), 1e-6, None)
    rr = (g["r"].to_numpy()/yb); ss = (g["s_win"].to_numpy() - 1.0)
    num_l = float((rr*ss).sum()); den_l = float((ss*ss).sum()) + eps
    lam_raw = num_l/den_l; n_l = len(g)
    lam = float(np.clip(lam_raw, -0.5, 0.5)) * (n_l/(n_l+L_LAMBDA)) if n_l>=MIN_N else 0.0
    rows_l.append((bid, lam))

gamma_table  = pd.DataFrame(rows_g, columns=["building_id","gamma"])
lambda_table = pd.DataFrame(rows_l, columns=["building_id","lambda"])

gamma_table.to_parquet(f"{DATA_DIR}/gamma_table.parquet", index=False)
lambda_table.to_parquet(f"{DATA_DIR}/lambda_table.parquet", index=False)
print("Saved: gamma_table.parquet, lambda_table.parquet")
```

```python
# Load artifacts we just saved (robust to re-runs)
alpha_table = pd.read_parquet(f"{DATA_DIR}/alpha_table_cf.parquet")
bias_table  = pd.read_parquet(f"{DATA_DIR}/bias_table_cf.parquet")
beta24      = pd.read_parquet(f"{DATA_DIR}/beta24_table.parquet")
beta168     = pd.read_parquet(f"{DATA_DIR}/beta168_table.parquet")
gamma_table = pd.read_parquet(f"{DATA_DIR}/gamma_table.parquet")
lambda_table= pd.read_parquet(f"{DATA_DIR}/lambda_table.parquet")

sub_hod = pd.read_csv(f"{DATA_DIR}/submission_hod.csv").rename(columns={"meter_reading":"meter_reading_hod"})
sub_ttm = pd.read_csv(f"{DATA_DIR}/submission_ttm.csv").rename(columns={"meter_reading":"meter_reading_ttm"})

# Targets with horizon + hod
targets = test_df.query("role=='target'")[["row_id","window_id","building_id","timestamp"]].copy()
targets = add_hoffset(targets); targets["hod"] = targets["timestamp"].dt.hour

# Input HoD and window mean for robust fallback
inp = test_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].copy().sort_values(["window_id","timestamp"])
yhod_inp_tab = (inp.assign(hod=inp["timestamp"].dt.hour).groupby(["window_id","hod"])["meter_reading"].mean().rename("y_hod_inp").reset_index())
wmean_inp = inp.groupby("window_id")["meter_reading"].mean().rename("wmean_inp").reset_index()

# Base blend
blend = (targets
         .merge(sub_hod, on="row_id", how="left")
         .merge(sub_ttm, on="row_id", how="left")
         .merge(alpha_table, on=["building_id","h_offset"], how="left")
         .merge(bias_table,  on=["building_id","h_offset"], how="left")
         .merge(yhod_inp_tab, on=["window_id","hod"], how="left")
         .merge(wmean_inp, on="window_id", how="left"))

y_hod_pred = pd.to_numeric(blend["meter_reading_hod"], errors="coerce")
y_hod_fallback = blend["y_hod_inp"].fillna(blend["wmean_inp"])
y_hod = y_hod_pred.fillna(y_hod_fallback)

y_ttm = pd.to_numeric(blend["meter_reading_ttm"], errors="coerce").fillna(y_hod)
blend["alpha"] = blend["alpha"].fillna(0.0); blend["bias"] = blend["bias"].fillna(0.0)
y_base = (blend["alpha"]*y_ttm + (1-blend["alpha"])*y_hod + blend["bias"]).clip(lower=0)

# Residual features z24/z168 from TEST inputs
idx_in = inp.set_index(["window_id","timestamp"])["meter_reading"].rename("mr_in")

lag24 = targets[["window_id","timestamp"]].copy()
lag24["timestamp_lag"] = lag24["timestamp"] - pd.Timedelta(hours=24)
lag24 = lag24.join(idx_in, on=["window_id","timestamp_lag"]).rename(columns={"mr_in":"lag24_val"})

lag168 = targets[["window_id","timestamp"]].copy()
lag168["timestamp_lag"] = lag168["timestamp"] - pd.Timedelta(hours=168)
lag168 = lag168.join(idx_in, on=["window_id","timestamp_lag"]).rename(columns={"mr_in":"lag168_val"})

blend = (blend.merge(lag24[["window_id","timestamp","lag24_val"]], on=["window_id","timestamp"], how="left")
              .merge(lag168[["window_id","timestamp","lag168_val"]], on=["window_id","timestamp"], how="left")
              .merge(beta24,  on=["building_id","h_offset"], how="left")
              .merge(beta168, on=["building_id","h_offset"], how="left"))

blend["beta24"] = blend["beta24"].fillna(0.0); blend["beta168"] = blend["beta168"].fillna(0.0)
y_hod_inp_safe = blend["y_hod_inp"].fillna(blend["wmean_inp"])
z24  = pd.to_numeric(blend["lag24_val"],  errors="coerce").fillna(y_hod_inp_safe) - y_hod_inp_safe
z168 = pd.to_numeric(blend["lag168_val"], errors="coerce").fillna(y_hod_inp_safe) - y_hod_inp_safe

# Safe residual cap
cap = 0.15 * y_base
y_corr = (y_base + (blend["beta24"]*z24 + blend["beta168"]*z168).clip(lower=-cap, upper=cap)).clip(lower=0)

# Window-level calibration (γ/λ)
w_inputs_te = (inp.groupby("window_id")
                  .agg(
                      head24_mean=("meter_reading", lambda s: s.head(24).mean()),
                      last24_mean=("meter_reading", lambda s: s.tail(24).mean()),
                      x_last=("meter_reading", "last"),
                  ).reset_index())
w_inputs_te["s_win"] = w_inputs_te["last24_mean"] / w_inputs_te["head24_mean"].clip(lower=1e-6)

mask0 = targets["h_offset"].to_numpy()==0
y0_df = targets.loc[mask0, ["window_id"]].copy()
y0_df["y0"] = np.asarray(y_corr)[mask0]
win_feat = w_inputs_te.merge(y0_df, on="window_id", how="left")
win_feat["y0"] = win_feat["y0"].fillna(win_feat["last24_mean"])
win_feat["delta_win"] = (win_feat["x_last"] - win_feat["y0"]).fillna(0.0)

out = (pd.DataFrame({"row_id": targets["row_id"]})
       .join(targets[["window_id","building_id","h_offset"]])
       .join(pd.DataFrame({"y_corr": y_corr}).set_index(targets.index)))

out = (out.merge(win_feat[["window_id","s_win","delta_win"]], on="window_id", how="left")
          .merge(gamma_table, on="building_id", how="left")
          .merge(lambda_table, on="building_id", how="left"))

out["s_win"]  = out["s_win"].fillna(1.0)
out["delta_win"] = out["delta_win"].fillna(0.0)
out["gamma"]  = out["gamma"].fillna(0.0)
out["lambda"] = out["lambda"].fillna(0.0)

scale = (1.0 + out["lambda"] * (out["s_win"] - 1.0)).clip(0.90, 1.15)
w_h = np.exp(-out["h_offset"]/12.0)
offs = out["gamma"] * out["delta_win"] * w_h

y_final = (out["y_corr"] * scale + offs)

# Final no-null guard
y_final = pd.to_numeric(y_final, errors="coerce").replace([np.inf, -np.inf], np.nan)
y_final = y_final.fillna(out["y_corr"]).fillna(y_base).fillna(0).clip(lower=0)

SUB_PATH = f"{DATA_DIR}/submission_final.csv"
pd.DataFrame({"row_id": targets["row_id"], "meter_reading": y_final}).sort_values("row_id").to_csv(SUB_PATH, index=False)
print("Wrote:", SUB_PATH, "| rows:", len(y_final))
```

```python
import pandas as pd, numpy as np
import os

DATA_DIR = "/content/drive/My Drive/IISC_data"

# Pick the latest prediction file you have
candidates = [
    f"{DATA_DIR}/submission_final_nonull.csv",
    f"{DATA_DIR}/submission_final.csv",
]
pred_path = None
for p in candidates:
    if os.path.exists(p):
        pred_path = p
        break
assert pred_path is not None, "No base submission file found."

pred = pd.read_csv(pred_path)  # columns: row_id, meter_reading
test_df = pd.read_csv(f"{DATA_DIR}/test.csv", parse_dates=["timestamp"])

# Join TEST targets to bring keys
def add_hoffset(df: pd.DataFrame) -> pd.DataFrame:
    out = df.sort_values(["window_id","timestamp"]).copy()
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

tgt_te = test_df.query("role=='target'")[["row_id","window_id","building_id","timestamp"]].copy()
tgt_te = add_hoffset(tgt_te)
tgt_te["dow"] = tgt_te["timestamp"].dt.dayofweek
tgt_te["is_weekend"] = (tgt_te["dow"] >= 5).astype(int)

df = tgt_te.merge(pred, on="row_id", how="left")
y = pd.to_numeric(df["meter_reading"], errors="coerce")

# Load artifacts
wknd = pd.read_parquet(f"{DATA_DIR}/wknd_table.parquet")
qbh = pd.read_parquet(f"{DATA_DIR}/quant_bhoff.parquet")
qb  = pd.read_parquet(f"{DATA_DIR}/quant_b.parquet")
with open(f"{DATA_DIR}/quant_global.txt","r") as f:
    qg01, qg99 = map(float, f.read().strip().split(","))

# --- Weekend nudge (additive) ---
df = df.merge(wknd, on=["building_id","h_offset"], how="left")
df["bias_weekend"] = df["bias_weekend"].fillna(0.0)
y1 = y + df["is_weekend"] * df["bias_weekend"]
y1 = pd.to_numeric(y1, errors="coerce")

# --- 24h smoothing within each window (gentle, horizon-dependent) ---
# Kernel and weight schedule
kernel = np.array([0.1, 0.2, 0.4, 0.2, 0.1])
w_sched = np.linspace(0.20, 0.55, 24).astype(float)  # more smoothing later
w_sched[0] = 0.10

# Apply per window_id
y2 = y1.copy().to_numpy()
for wid, idx in df.groupby("window_id").indices.items():
    # idx are positions in df for this window's 24 target rows (already sorted by timestamp from add_hoffset)
    g = y1.iloc[idx].to_numpy()
    # reflect-pad for edges
    ext = np.r_[g[1:2][::-1], g, g[-2:-1][::-1]]  # pad 1 on each side
    sm  = np.convolve(ext, kernel, mode="same")[1:-1]
    # blend with schedule
    w = w_sched[:len(sm)]
    y2[idx] = (1.0 - w) * g + w * sm

y2 = np.clip(y2, 0, None)

# --- Percentile clip using TRAIN quantiles with fallbacks ---
# Merge (building,h_offset) quantiles
df = df.merge(qbh, on=["building_id","h_offset"], how="left")
df = df.merge(qb, on="building_id", how="left")

# Cascade fallbacks
qlo = df["q01"].fillna(df["q01_b"]).fillna(qg01)
qhi = df["q99"].fillna(df["q99_b"]).fillna(qg99)

# Slightly relaxed band to avoid over-clipping
band_lo = 0.95 * qlo
band_hi = 1.05 * qhi

y3 = np.clip(y2, band_lo.to_numpy(), band_hi.to_numpy())
y3 = np.nan_to_num(y3, nan=0.0, posinf=0.0, neginf=0.0)

# --- Final write (no NaNs, >=0) ---
sub_out = pd.DataFrame({
    "row_id": df["row_id"].to_numpy(),
    "meter_reading": np.clip(y3, 0, None)
}).sort_values("row_id")

assert sub_out["meter_reading"].notna().all()
assert (sub_out["meter_reading"] >= 0).all()
assert len(sub_out) == 84600 and sub_out["row_id"].is_unique

OUT = f"{DATA_DIR}/submission_postproc_weekend_smooth_clip.csv"
sub_out.to_csv(OUT, index=False)
print("Wrote:", OUT, "| rows:", len(sub_out))
```

```python
import os, numpy as np, pandas as pd

DATA_DIR = "/content/drive/My Drive/IISC_data"

# Base submissions (add/remove if you have more)
cands = {
    "final":  f"{DATA_DIR}/submission_final_nonull.csv",   # 0.43723
    "calib":  f"{DATA_DIR}/submission_calibrated.csv",     # 0.43820
    "lag24":  f"{DATA_DIR}/submission_blend_cf_lag24.csv", # 0.44575
    "ttm":    f"{DATA_DIR}/submission_ttm.csv",
    "hod":    f"{DATA_DIR}/submission_hod.csv",
    # keep median_top out of inputs to avoid double-counting
}

# Load existing
dfs = {}
for k, p in cands.items():
    if os.path.exists(p):
        df = pd.read_csv(p)
        if set(df.columns)=={"row_id","meter_reading"}:
            dfs[k] = df.rename(columns={"meter_reading": f"m_{k}"}).sort_values("row_id")
            print(f"Loaded: {k} ({len(df)})")
assert "final" in dfs, "Need submission_final_nonull.csv present."

# Merge on row_id
base = dfs["final"][["row_id"]].copy()
for k, df in dfs.items():
    base = base.merge(df, on="row_id", how="left")

cols = [c for c in ["m_final","m_calib","m_lag24","m_ttm","m_hod"] if c in base]

def row_trimmed_mean(mat_row):
    v = mat_row[np.isfinite(mat_row)]
    if v.size == 0: return np.nan
    v = np.sort(v)
    if v.size >= 5:
        v = v[1:-1]                # drop min & max
    elif v.size == 4:
        v = v[1:-1]                # drop extremes → mean middle two
    elif v.size == 3:
        return float(np.median(v)) # median for 3
    return float(np.mean(v))

def row_winsor_mean(mat_row):
    v = mat_row[np.isfinite(mat_row)]
    if v.size == 0: return np.nan
    v = np.sort(v)
    if v.size >= 3:
        v[0]  = v[1]               # winsorize ends
        v[-1] = v[-2]
    return float(np.mean(v))

def row_power_mean(mat_row, p=0.5):
    v = mat_row[np.isfinite(mat_row)]
    if v.size == 0: return np.nan
    if p == 0:  # geometric mean limit
        v = np.clip(v, 1e-9, None)
        return float(np.exp(np.mean(np.log(v))))
    return float((np.mean(np.power(np.clip(v, 0, None), p))) ** (1.0/p))

M = base[cols].to_numpy(float)

trimmed = np.array([row_trimmed_mean(M[i]) for i in range(M.shape[0])])
winsor  = np.array([row_winsor_mean(M[i])  for i in range(M.shape[0])])
p05     = np.array([row_power_mean(M[i], p=0.5) for i in range(M.shape[0])])

def write_safe(name, vec):
    y = pd.to_numeric(pd.Series(vec), errors="coerce").replace([np.inf,-np.inf], np.nan)
    # Fallback chain: ensemble → m_final → 0
    y = y.fillna(base.get("m_final")).fillna(0).clip(lower=0)
    out = base[["row_id"]].copy()
    out["meter_reading"] = y
    path = f"{DATA_DIR}/{name}.csv"
    out.to_csv(path, index=False)
    print("Wrote:", path, "| rows:", len(out))
    return path

p1 = write_safe("submission_ens_trimmed.csv", trimmed)
p2 = write_safe("submission_ens_winsor.csv",  winsor)
p3 = write_safe("submission_ens_power05.csv", p05)

# Also try a convex blend between your best median and final (if median file exists)
med_path = f"{DATA_DIR}/submission_ens_median_top.csv"
if os.path.exists(med_path):
    med = pd.read_csv(med_path).sort_values("row_id")["meter_reading"].to_numpy()
    f   = base["m_final"].to_numpy()
    p4 = write_safe("submission_ens_medfin_60_40.csv", 0.60*med + 0.40*f)
    p5 = write_safe("submission_ens_medfin_70_30.csv", 0.70*med + 0.30*f)
```

```python
def build_window_features(inputs_window: pd.DataFrame):
    """Return dict with per-window scalar features computed from the last 168h."""
    s = inputs_window["meter_reading"].to_numpy(dtype=float)
    # Safety for short windows (shouldn't happen)
    if len(s) < 168:
        s = np.pad(s, (168-len(s), 0), mode="edge")
    # Scalars from the entire window
    feats = {
        "x_last": float(s[-1]),
        "m3":  float(s[-3:].mean()),
        "m6":  float(s[-6:].mean()),
        "m12": float(s[-12:].mean()),
        "m24": float(s[-24:].mean()),
        "m48": float(s[-48:].mean()),
        "std24": float(s[-24:].std(ddof=0)),
        "min168": float(s.min()),
        "max168": float(s.max()),
        "ratio_last24_first24": float(s[-24:].mean() / max(s[:24].mean(), 1e-6)),
        "slope48": float(np.polyfit(np.arange(48), s[-48:], 1)[0]) if np.all(np.isfinite(s[-48:])) else 0.0,
    }
    # Hour-of-day mean over the 7 days (168 = 7*24)
    hod_map = inputs_window.assign(hod=inputs_window["timestamp"].dt.hour)\
                           .groupby("hod")["meter_reading"].mean().to_dict()
    return feats, hod_map, s

def build_features_frame(inputs_all: pd.DataFrame, targets_all: pd.DataFrame, is_train: bool):
    """Construct one row per target with features from its window inputs."""
    rows = []
    # Pre-chunk targets by window for speed
    t_groups = {wid: g.copy() for wid, g in targets_all.groupby("window_id")}
    for wid, g_inp in inputs_all.groupby("window_id"):
        feats, hod_map, s = build_window_features(g_inp)
        if wid not in t_groups:
            continue
        g_tgt = t_groups[wid].sort_values("timestamp").copy()
        # Make sure we have horizons (0..23)
        if "h_offset" not in g_tgt.columns:
            g_tgt = add_hoffset(g_tgt)
        for _, r in g_tgt.iterrows():
            h = int(r["h_offset"])
            hod = int(r["timestamp"].hour)
            week_ago = float(s[h])         # same hour last week (aligns inside 168h)
            hod_mean = float(hod_map.get(hod, np.nan))
            rows.append({
                # keys
                **({"row_id": int(r["row_id"])} if not is_train else {}),
                "window_id": r["window_id"],
                "building_id": r["building_id"],
                "h_offset": h,
                # targets (train)
                **({"ytrue": float(r["ytrue"])} if is_train else {}),
                # scalar window feats
                **feats,
                # horizon/weekly alignment feats
                "week_ago": week_ago,
                "hod_mean": hod_mean,
                "week_ago_minus_hod": week_ago - (hod_mean if np.isfinite(hod_mean) else week_ago),
                # cyclical time
                "hod": hod,
                "sin_h": np.sin(2*np.pi*hod/24.0),
                "cos_h": np.cos(2*np.pi*hod/24.0),
            })
    df = pd.DataFrame(rows)
    # Robust fill
    for c in ["hod_mean","week_ago_minus_hod"]:
        if c in df.columns:
            df[c] = df[c].fillna(df[c].median())
    # Building as category
    df["building_id"] = df["building_id"].astype("category")
    return df

# Build frames
feat_tr = build_features_frame(inp_tr, tgt_tr, is_train=True)
feat_te = build_features_frame(inp_te, tgt_te, is_train=False)
print("Train feats:", feat_tr.shape, " Test feats:", feat_te.shape)
```

```python
# Rebuild LGBM TRAIN predictions (pred_tr) and TEST preds (submission_lgbm.csv)

import pandas as pd, numpy as np
from sklearn.preprocessing import StandardScaler
from lightgbm import LGBMRegressor

DATA_DIR = "/content/drive/My Drive/IISC_data"

# 1) Reload data
train_df = pd.read_csv(f"{DATA_DIR}/train.csv", parse_dates=["timestamp"])
test_df  = pd.read_csv(f"{DATA_DIR}/test.csv",  parse_dates=["timestamp"])

def add_hoffset(df: pd.DataFrame) -> pd.DataFrame:
    out = df.sort_values(["window_id","timestamp"]).copy()
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

# 2) Feature builders (same as earlier)
def build_window_features(inputs_window: pd.DataFrame):
    s = inputs_window["meter_reading"].to_numpy(float)
    if len(s) < 168:
        s = np.pad(s, (168-len(s), 0), mode="edge")
    feats = {
        "x_last": float(s[-1]),
        "m3":  float(s[-3:].mean()),
        "m6":  float(s[-6:].mean()),
        "m12": float(s[-12:].mean()),
        "m24": float(s[-24:].mean()),
        "m48": float(s[-48:].mean()),
        "std24": float(s[-24:].std(ddof=0)),
        "min168": float(s.min()),
        "max168": float(s.max()),
        "ratio_last24_first24": float(s[-24:].mean() / max(s[:24].mean(), 1e-6)),
        "slope48": float(np.polyfit(np.arange(48), s[-48:], 1)[0]),
    }
    hod_map = (inputs_window.assign(hod=inputs_window["timestamp"].dt.hour)
               .groupby("hod")["meter_reading"].mean().to_dict())
    return feats, hod_map, s

def build_features_frame(inputs_all: pd.DataFrame, targets_all: pd.DataFrame, is_train: bool):
    rows = []
    t_groups = {wid: g.copy() for wid, g in targets_all.groupby("window_id")}
    for wid, g_inp in inputs_all.groupby("window_id"):
        feats, hod_map, s = build_window_features(g_inp.sort_values("timestamp"))
        if wid not in t_groups:
            continue
        g_tgt = add_hoffset(t_groups[wid].sort_values("timestamp"))
        for _, r in g_tgt.iterrows():
            h = int(r["h_offset"]); hod = int(r["timestamp"].hour)
            week_ago = float(s[h])  # align within 168h
            hod_mean = float(hod_map.get(hod, week_ago))
            rows.append({
                **({"row_id": int(r["row_id"])} if not is_train else {}),
                "window_id": r["window_id"],
                "building_id": r["building_id"],
                "h_offset": h,
                **({"ytrue": float(r["ytrue"])} if is_train else {}),
                **feats,
                "week_ago": week_ago,
                "hod_mean": hod_mean,
                "week_ago_minus_hod": week_ago - hod_mean,
                "hod": hod,
                "sin_h": np.sin(2*np.pi*hod/24.0),
                "cos_h": np.cos(2*np.pi*hod/24.0),
            })
    df = pd.DataFrame(rows)
    df["building_id"] = df["building_id"].astype("category")
    for c in ["hod_mean","week_ago_minus_hod"]:
        if c in df: df[c] = df[c].fillna(df[c].median())
    return df

# 3) Build frames
tgt_tr = add_hoffset(train_df.query("role=='target'")[["window_id","building_id","timestamp","meter_reading"]]
                     .rename(columns={"meter_reading":"ytrue"}))
tgt_te = add_hoffset(test_df.query("role=='target'")[["row_id","window_id","building_id","timestamp"]])
inp_tr = train_df.query("role=='input'")[["window_id","building_id","timestamp","meter_reading"]].sort_values(["window_id","timestamp"])
inp_te = test_df.query("role=='input'")[["window_id","building_id","timestamp","meter_reading"]].sort_values(["window_id","timestamp"])

feat_tr_lgb = build_features_frame(inp_tr, tgt_tr, is_train=True)
feat_te_lgb = build_features_frame(inp_te, tgt_te, is_train=False)

# 4) Train 24 LGBM models; get TRAIN preds (pred_tr) and TEST preds (submission_lgbm.csv)
drop_cols = {"row_id","window_id","ytrue"}
X_cols = [c for c in feat_tr_lgb.columns if c not in drop_cols]
params = dict(
    n_estimators=800, learning_rate=0.03, num_leaves=63,
    min_data_in_leaf=60, subsample=0.9, colsample_bytree=0.8,
    reg_alpha=0.1, reg_lambda=1.0, random_state=42, n_jobs=-1, verbosity=-1,
)

from lightgbm import LGBMRegressor
pred_tr_list, pred_te_list = [], []
for h in range(24):
    tr_h = feat_tr_lgb[feat_tr_lgb["h_offset"]==h].copy()
    te_h = feat_te_lgb[feat_te_lgb["h_offset"]==h].copy()

    X_tr = tr_h[X_cols]; y_tr = tr_h["ytrue"].astype(float)
    X_te = te_h[X_cols]

    cont_cols = [c for c in X_cols if c not in ["building_id"]]
    scaler = StandardScaler().fit(X_tr[cont_cols])
    X_tr_scaled = X_tr.copy(); X_tr_scaled[cont_cols] = scaler.transform(X_tr[cont_cols])
    X_te_scaled = X_te.copy(); X_te_scaled[cont_cols] = scaler.transform(X_te[cont_cols])

    model = LGBMRegressor(**params)
    model.fit(X_tr_scaled, y_tr, categorical_feature=["building_id"])

    # TRAIN predictions for super-stack learning
    yhat_tr = np.clip(model.predict(X_tr_scaled), 0, None)
    out_tr = tr_h[["window_id","h_offset"]].copy(); out_tr["yhat_lgbm"] = yhat_tr
    pred_tr_list.append(out_tr)

    # TEST predictions for stacking later (if needed)
    yhat_te = np.clip(model.predict(X_te_scaled), 0, None)
    out_te = te_h[["row_id"]].copy(); out_te["meter_reading_lgbm"] = yhat_te
    pred_te_list.append(out_te)

pred_tr = pd.concat(pred_tr_list, ignore_index=True)
pred_te = pd.concat(pred_te_list, ignore_index=True).sort_values("row_id")

pred_tr.to_parquet(f"{DATA_DIR}/pred_tr_lgbm.parquet", index=False)
pred_te.to_csv(f"{DATA_DIR}/submission_lgbm.csv", index=False)

print("Rebuilt pred_tr:", pred_tr.shape, "| Wrote submission_lgbm.csv:", pred_te.shape)
```

```python
import pandas as pd, numpy as np
from tqdm import tqdm
import torch
from tsfm_public import TimeSeriesPreprocessor, TinyTimeMixerForPrediction, TimeSeriesForecastingPipeline

DATA_DIR = "/content/drive/My Drive/IISC_data"

train_df = pd.read_csv(f"{DATA_DIR}/train.csv", parse_dates=["timestamp"])
test_df  = pd.read_csv(f"{DATA_DIR}/test.csv",  parse_dates=["timestamp"])

def add_hoffset(df):
    out = df.sort_values(["window_id","timestamp"]).copy()
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

# TRAIN targets
tgt_tr = (train_df.query("role=='target'")
          [["window_id","building_id","timestamp","meter_reading"]]
          .rename(columns={"meter_reading":"ytrue"}))
tgt_tr = add_hoffset(tgt_tr); tgt_tr["hod"] = tgt_tr["timestamp"].dt.hour

# HoD on TRAIN
inp_tr = train_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].copy()
inp_tr["hod"] = inp_tr["timestamp"].dt.hour
hod_tab = (inp_tr.groupby(["window_id","hod"])["meter_reading"].mean().rename("y_hod").reset_index())
wmean_tr = (inp_tr.groupby("window_id")["meter_reading"].mean().rename("wmean").reset_index())

feat_tr = (tgt_tr.merge(hod_tab, on=["window_id","hod"], how="left")
                 .merge(wmean_tr, on="window_id", how="left"))
feat_tr["y_hod"] = feat_tr["y_hod"].fillna(feat_tr["wmean"]).astype(float)

# LGBM train preds (from your earlier loop). If missing, abort.
assert 'pred_tr' in globals(), "Run the LGBM training cell first to populate pred_tr."
feat_tr = feat_tr.merge(pred_tr.rename(columns={"yhat_lgbm":"y_lgbm"}),
                        on=["window_id","h_offset"], how="left")

# TTM zero-shot on TRAIN (all windows, efficient)
tsp_tr = TimeSeriesPreprocessor(
    timestamp_column="timestamp",
    id_columns=["building_id","window_id"],
    target_columns=["meter_reading"],
    control_columns=[],
    context_length=180, forecast_length=60,
    scaling=True, scaler_type="standard", freq="h",
)
tsp_tr.train(train_df.query("role=='input'"))

ttm_tr = TinyTimeMixerForPrediction.from_pretrained(
    "ibm-granite/granite-timeseries-ttm-r2",
    revision="180-60-ft-l1-r2.1",
    num_input_channels=1,
).eval()
device = "cuda" if torch.cuda.is_available() else "cpu"
pipe_tr = TimeSeriesForecastingPipeline(ttm_tr, explode_forecasts=True, feature_extractor=tsp_tr, device=device)

def pad_to_180(win_df, ts_col="timestamp"):
    win = win_df.sort_values(ts_col).copy()
    if len(win) == 168:
        pad = win.iloc[:12].copy()
        pad[ts_col] = pad[ts_col] - pd.to_timedelta(np.arange(12, 0, -1), unit="h")
        win = pd.concat([pad, win], ignore_index=True).sort_values(ts_col)
    return win

preds = []
for wi, g in tqdm(train_df.groupby("window_id", sort=False), total=train_df["window_id"].nunique(), desc="TTM train"):
    inp = g[g["role"]=="input"].copy()
    tar = g[g["role"]=="target"][["timestamp"]].copy()
    p = pipe_tr(pad_to_180(inp)).iloc[:24][["meter_reading"]].reset_index(drop=True)
    tar = tar.reset_index(drop=True); tar["y_ttm"] = p["meter_reading"].astype(float); tar["window_id"] = wi
    preds.append(tar)
pred_ttm_tr = pd.concat(preds, ignore_index=True)

feat_tr = (feat_tr.merge(pred_ttm_tr, on=["window_id","timestamp"], how="left")
                   .rename(columns={"y_ttm":"y_ttm"}))
# Final clean
feat_tr["y_lgbm"] = pd.to_numeric(feat_tr["y_lgbm"], errors="coerce")
feat_tr["y_ttm"]  = pd.to_numeric(feat_tr["y_ttm"],  errors="coerce")
feat_tr["ytrue"]  = pd.to_numeric(feat_tr["ytrue"], errors="coerce")
print("TRAIN meta-frame:", feat_tr.shape, "NaNs:", feat_tr.isna().sum().to_dict())
```

```python
import numpy as np, pandas as pd

# Fill occasional holes for stability
feat_tr["y_lgbm"] = feat_tr["y_lgbm"].fillna(feat_tr["y_hod"])
feat_tr["y_ttm"]  = feat_tr["y_ttm"].fillna(feat_tr["y_hod"])

lam = 200.0  # ridge strength
eps = 1e-9

rows = []
for (bid, hoff), g in feat_tr.groupby(["building_id","h_offset"], sort=False):
    X = np.column_stack([
        g["y_hod"].to_numpy(float),
        g["y_ttm"].to_numpy(float),
        g["y_lgbm"].to_numpy(float),
        np.ones(len(g), float)  # bias
    ])
    y = g["ytrue"].to_numpy(float)
    XtX = X.T @ X
    XtX[:3, :3] += lam * np.eye(3)  # penalize only model weights (not bias)
    w = np.linalg.solve(XtX + eps*np.eye(4), X.T @ y)
    # nonneg + renorm for first 3 weights
    w[:3] = np.clip(w[:3], 0, None)
    s = w[:3].sum()
    if s > 0:
        w[:3] = w[:3] / s
    rows.append((bid, hoff, *w))

w_bh = pd.DataFrame(rows, columns=["building_id","h_offset","w_hod","w_ttm","w_lgbm","bias"])

# Smooth weights along horizons per building
ker = np.array([0.2, 0.6, 0.2])
def smooth_group(df, cols):
    out = df.copy()
    for c in cols:
        v = out[c].to_numpy()
        out[c] = np.convolve(v, ker, mode="same")
    return out

w_bh = (w_bh.sort_values(["building_id","h_offset"])
             .groupby("building_id", group_keys=False)
             .apply(lambda d: smooth_group(d, ["w_hod","w_ttm","w_lgbm","bias"])))
# Re-normalize after smoothing
s = w_bh[["w_hod","w_ttm","w_lgbm"]].sum(axis=1)
for c in ["w_hod","w_ttm","w_lgbm"]:
    w_bh[c] = np.where(s>0, w_bh[c]/s, 1.0/3)
w_bh["bias"] = w_bh["bias"].fillna(0.0)

# Fallback weights per building (averaged across horizons)
w_b = (w_bh.groupby("building_id")[["w_hod","w_ttm","w_lgbm","bias"]].mean()
       .reset_index().rename(columns={c:f"{c}_b" for c in ["w_hod","w_ttm","w_lgbm","bias"]}))

w_bh.to_parquet(f"{DATA_DIR}/stack_weights_bh.parquet", index=False)
w_b.to_parquet(f"{DATA_DIR}/stack_weights_b.parquet", index=False)
print("Saved weights:", w_bh.shape, w_b.shape)
```

```python
# Load TEST base predictions you already have
sub_hod = pd.read_csv(f"{DATA_DIR}/submission_hod.csv").rename(columns={"meter_reading":"y_hod"})
sub_ttm = pd.read_csv(f"{DATA_DIR}/submission_ttm.csv").rename(columns={"meter_reading":"y_ttm"})
sub_lgb = pd.read_csv(f"{DATA_DIR}/submission_lgbm.csv").rename(columns={"meter_reading_lgbm":"y_lgbm"})

targets = test_df.query("role=='target'")[["row_id","window_id","building_id","timestamp"]].copy()
targets = add_hoffset(targets)

# Join predictions
S = (targets.merge(sub_hod, on="row_id", how="left")
            .merge(sub_ttm, on="row_id", how="left")
            .merge(sub_lgb, on="row_id", how="left"))

# Fill safety
S["y_ttm"]  = pd.to_numeric(S["y_ttm"], errors="coerce")
S["y_lgbm"] = pd.to_numeric(S["y_lgbm"], errors="coerce")
S["y_hod"]  = pd.to_numeric(S["y_hod"], errors="coerce")
S["y_ttm"]  = S["y_ttm"].fillna(S["y_hod"])
S["y_lgbm"] = S["y_lgbm"].fillna(S["y_hod"])

# Merge weights (bh first, then building fallback)
w_bh = pd.read_parquet(f"{DATA_DIR}/stack_weights_bh.parquet")
w_b  = pd.read_parquet(f"{DATA_DIR}/stack_weights_b.parquet")
S = (S.merge(w_bh, on=["building_id","h_offset"], how="left")
       .merge(w_b, on=["building_id"], how="left"))

for c in ["w_hod","w_ttm","w_lgbm","bias"]:
    S[c] = S[c].fillna(S[f"{c}_b"]).fillna(1.0/3 if c!="bias" else 0.0)

# Main super-stack
y_ss = (S["w_hod"]*S["y_hod"] + S["w_ttm"]*S["y_ttm"] + S["w_lgbm"]*S["y_lgbm"] + S["bias"]).clip(lower=0)

# (Optional) gentle calibration with your γ/λ if you already learned them
import os
y_final_main = y_ss.copy()
gamma_p = f"{DATA_DIR}/gamma_table.parquet"
lambda_p= f"{DATA_DIR}/lambda_table.parquet"
if os.path.exists(gamma_p) and os.path.exists(lambda_p):
    gamma_table  = pd.read_parquet(gamma_p)
    lambda_table = pd.read_parquet(lambda_p)
    # window features
    inp_te = test_df.query("role=='input'")[["window_id","timestamp","meter_reading"]].sort_values(["window_id","timestamp"])
    w_inputs_te = (inp_te.groupby("window_id")
                   .agg(head24_mean=("meter_reading", lambda s: s.head(24).mean()),
                        last24_mean=("meter_reading", lambda s: s.tail(24).mean()),
                        x_last=("meter_reading", "last"))
                   .reset_index())
    w_inputs_te["s_win"] = w_inputs_te["last24_mean"] / w_inputs_te["head24_mean"].clip(lower=1e-6)
    mask0 = S["h_offset"].to_numpy()==0
    y0_df = S.loc[mask0, ["window_id"]].copy()
    y0_df["y0"] = y_ss[mask0].to_numpy()
    win_feat = w_inputs_te.merge(y0_df, on="window_id", how="left")
    win_feat["y0"] = win_feat["y0"].fillna(win_feat["last24_mean"])
    win_feat["delta_win"] = (win_feat["x_last"] - win_feat["y0"]).fillna(0.0)

    D = (S[["row_id","window_id","building_id","h_offset"]]
         .merge(win_feat[["window_id","s_win","delta_win"]], on="window_id", how="left")
         .merge(gamma_table, on="building_id", how="left")
         .merge(lambda_table, on="building_id", how="left"))
    D["s_win"]  = D["s_win"].fillna(1.0)
    D["delta_win"] = D["delta_win"].fillna(0.0)
    D["gamma"]  = D["gamma"].fillna(0.0)
    D["lambda"] = D["lambda"].fillna(0.0)
    scale = (1.0 + D["lambda"] * (D["s_win"] - 1.0)).clip(0.90, 1.15)
    w_h = np.exp(-D["h_offset"]/12.0)
    offs = D["gamma"] * D["delta_win"] * w_h
    y_final_main = (y_ss * scale + offs).clip(lower=0)

# Safe variant: tighter scale caps
y_final_safe = y_final_main.copy()
if 'scale' in locals():
    scale_safe = scale.clip(0.92, 1.10)
    y_final_safe = (y_ss * scale_safe + offs).clip(lower=0)

# Write files
OUT1 = f"{DATA_DIR}/submission_superstack.csv"
OUT2 = f"{DATA_DIR}/submission_superstack_safe.csv"
pd.DataFrame({"row_id": S["row_id"], "meter_reading": y_final_main}).sort_values("row_id").to_csv(OUT1, index=False)
pd.DataFrame({"row_id": S["row_id"], "meter_reading": y_final_safe}).sort_values("row_id").to_csv(OUT2, index=False)
print("Wrote:", OUT1, "and", OUT2)
```

```python
import pandas as pd, numpy as np

DATA_DIR = "/content/drive/My Drive/IISC_data"

# 1) TRAIN target frame with horizon + HoD and TTM (already computed in your meta-frame step)
#    feat_tr has columns: [building_id, window_id, timestamp, h_offset, ytrue, y_hod]
#    pred_ttm_tr has ['window_id','timestamp','y_ttm']
assert 'feat_tr' in globals(), "Run the meta-frame step so feat_tr exists."
assert 'pred_ttm_tr' in globals(), "Run the TTM train loop so pred_ttm_tr exists."

feat_tr_blend = (feat_tr[["building_id","window_id","h_offset","timestamp","ytrue","y_hod"]]
                 .merge(pred_ttm_tr, on=["window_id","timestamp"], how="left")
                 .rename(columns={"y_ttm":"y_ttm_tr"}))

# LGBM train predictions (you rebuilt earlier as pred_tr)
if 'pred_tr' not in globals():
    pred_tr = pd.read_parquet(f"{DATA_DIR}/pred_tr_lgbm.parquet")
feat_tr_blend = feat_tr_blend.merge(pred_tr, on=["window_id","h_offset"], how="left") \
                             .rename(columns={"yhat_lgbm":"y_lgbm_tr"})

# 2) Alpha & bias tables (closed-form) – already computed earlier
alpha_table = pd.read_parquet(f"{DATA_DIR}/alpha_table_cf.parquet")
bias_table  = pd.read_parquet(f"{DATA_DIR}/bias_table_cf.parquet")

feat_tr_blend = (feat_tr_blend
                 .merge(alpha_table, on=["building_id","h_offset"], how="left")
                 .merge(bias_table,  on=["building_id","h_offset"], how="left"))
feat_tr_blend["alpha"] = feat_tr_blend["alpha"].fillna(0.0)
feat_tr_blend["bias"]  = feat_tr_blend["bias"].fillna(0.0)

# 3) Base prediction on TRAIN using HoD+TTM+bias; fill any TTM gaps with HoD
feat_tr_blend["y_ttm_tr"] = pd.to_numeric(feat_tr_blend["y_ttm_tr"], errors="coerce").fillna(feat_tr_blend["y_hod"])
feat_tr_blend["y_base_tr"] = (feat_tr_blend["alpha"]*feat_tr_blend["y_ttm_tr"]
                              + (1.0 - feat_tr_blend["alpha"])*feat_tr_blend["y_hod"]
                              + feat_tr_blend["bias"]).clip(lower=0)

# 4) Closed-form optimal weight w between LGBM and base, per (building,h_offset) with shrinkage
#    Min E[(w*d - r)^2], where d = (y_lgbm - y_base), r = (y_base - ytrue)*(-1)
feat_tr_blend["d"] = (pd.to_numeric(feat_tr_blend["y_lgbm_tr"], errors="coerce")
                      - feat_tr_blend["y_base_tr"])
feat_tr_blend["r"] = (feat_tr_blend["ytrue"] - feat_tr_blend["y_base_tr"])

rows = []
LAM = 50.0    # ridge on denominator (stability)
SHR = 200.0   # shrinkage strength toward 0 (prefer base)
for (bid, hoff), g in feat_tr_blend.dropna(subset=["d","r"]).groupby(["building_id","h_offset"], sort=False):
    num = float((g["d"] * g["r"]).sum())
    den = float((g["d"] * g["d"]).sum() + LAM)
    w_raw = num / den
    # clip to reasonable range; we don't want to overweight LGBM
    w_raw = float(np.clip(w_raw, 0.0, 0.6))
    n = len(g)
    w_shr = w_raw * (n / (n + SHR))      # shrink toward 0 (base)
    rows.append((bid, hoff, w_shr, n))

w_bh = pd.DataFrame(rows, columns=["building_id","h_offset","w_lgbm","n"])

# Smooth weights across horizons within each building
ker = np.array([0.25, 0.5, 0.25])
def smooth_group(df):
    out = df.sort_values("h_offset").copy()
    v = out["w_lgbm"].to_numpy()
    out["w_lgbm"] = np.convolve(v, ker, mode="same")
    return out

w_bh = w_bh.groupby("building_id", group_keys=False).apply(smooth_group)
# Fallback per building (mean across horizons)
w_b = w_bh.groupby("building_id")["w_lgbm"].mean().rename("w_lgbm_b").reset_index()

w_bh.to_parquet(f"{DATA_DIR}/stack_w_bh.parquet", index=False)
w_b.to_parquet(f"{DATA_DIR}/stack_w_b.parquet", index=False)
print("Saved weights:", w_bh.shape, w_b.shape, "w range:", (w_bh['w_lgbm'].min(), w_bh['w_lgbm'].max()))
```

```python
# --- Isotonic meta-calibration on TRAIN (fixed for string building_id) ---
import pandas as pd, numpy as np
from sklearn.isotonic import IsotonicRegression

DATA_DIR = "/content/drive/My Drive/IISC_data"

alpha_table = pd.read_parquet(f"{DATA_DIR}/alpha_table_cf.parquet")
bias_table  = pd.read_parquet(f"{DATA_DIR}/bias_table_cf.parquet")

# Prereqs:
# feat_tr (has columns: building_id, window_id, timestamp, h_offset, ytrue, y_hod)
# pred_ttm_tr: ['window_id','timestamp','y_ttm']
# pred_tr:     ['window_id','h_offset','yhat_lgbm']
assert 'feat_tr' in globals(), "feat_tr missing. Run your meta-frame step first."

# Fallbacks if not in memory
if 'pred_ttm_tr' not in globals():
    print("Warning: pred_ttm_tr missing; using HoD as TTM fallback.")
    pred_ttm_tr = feat_tr[["window_id","timestamp","y_hod"]].rename(columns={"y_hod":"y_ttm"})
if 'pred_tr' not in globals():
    pred_tr = pd.read_parquet(f"{DATA_DIR}/pred_tr_lgbm.parquet")

# Keep building_id as STRING everywhere
feat_tr["building_id"] = feat_tr["building_id"].astype(str)
pred_tr = pred_tr.copy(); pred_tr["h_offset"] = pred_tr["h_offset"].astype(int)

W_LGBM = 0.20  # match in TEST

t = (feat_tr[["building_id","window_id","h_offset","timestamp","ytrue","y_hod"]]
     .merge(pred_ttm_tr.rename(columns={"y_ttm":"y_ttm_tr"}), on=["window_id","timestamp"], how="left")
     .merge(pred_tr.rename(columns={"yhat_lgbm":"y_lgbm_tr"}), on=["window_id","h_offset"], how="left")
     .merge(alpha_table, on=["building_id","h_offset"], how="left")
     .merge(bias_table,  on=["building_id","h_offset"], how="left"))

t[["alpha","bias"]] = t[["alpha","bias"]].fillna(0.0)
t["y_ttm_tr"] = pd.to_numeric(t["y_ttm_tr"], errors="coerce").fillna(t["y_hod"])
t["y_lgbm_tr"]= pd.to_numeric(t["y_lgbm_tr"],errors="coerce").fillna(t["y_hod"])

t["y_base_tr"] = (t["alpha"]*t["y_ttm_tr"] + (1.0-t["alpha"])*t["y_hod"] + t["bias"]).clip(lower=0)
t["y_pred_tr"] = np.clip((1.0-W_LGBM)*t["y_base_tr"] + W_LGBM*t["y_lgbm_tr"], 0, None)

# --- Fit isotonic per (building,h_offset) with shrinkage + fallbacks ---
MIN_N_BH = 180
MIN_N_B   = 800
SHR_K     = 600

calib_bh = {}  # key: (building_id:str, h:int) -> (iso_model, rho)
calib_b  = {}  # key: building_id:str -> (iso_model, rho)

iso_global = IsotonicRegression(out_of_bounds="clip")
mask = t[["y_pred_tr","ytrue"]].dropna().index
iso_global.fit(t.loc[mask,"y_pred_tr"], t.loc[mask,"ytrue"])

for (bid, hoff), g in t.dropna(subset=["y_pred_tr","ytrue"]).groupby(["building_id","h_offset"], sort=False):
    n = len(g)
    if n >= MIN_N_BH:
        ir = IsotonicRegression(out_of_bounds="clip")
        ir.fit(g["y_pred_tr"].to_numpy(), g["ytrue"].to_numpy())
        rho = n / (n + SHR_K)
        calib_bh[(str(bid), int(hoff))] = (ir, float(rho))

for bid, g in t.dropna(subset=["y_pred_tr","ytrue"]).groupby("building_id", sort=False):
    n = len(g)
    if n >= MIN_N_B:
        ir = IsotonicRegression(out_of_bounds="clip")
        ir.fit(g["y_pred_tr"].to_numpy(), g["ytrue"].to_numpy())
        rho = n / (n + SHR_K)
        calib_b[str(bid)] = (ir, float(rho))

# Quantile bands for later clipping
q_bh = (t.groupby(["building_id","h_offset"])["ytrue"].quantile([0.01,0.99]).unstack(-1)
          .rename(columns={0.01:"q01",0.99:"q99"}).reset_index())
q_b  = (t.groupby("building_id")["ytrue"].quantile([0.01,0.99]).unstack(-1)
          .rename(columns={0.01:"q01_b",0.99:"q99_b"}).reset_index())
qg01, qg99 = float(t["ytrue"].quantile(0.01)), float(t["ytrue"].quantile(0.99))

print("Calibrators:", len(calib_bh), "per-(building,h),", len(calib_b), "per-building, global OK.")
```

```python
import pandas as pd, numpy as np

DATA_DIR = "/content/drive/My Drive/IISC_data"

test_df  = pd.read_csv(f"{DATA_DIR}/test.csv",  parse_dates=["timestamp"])
def add_hoffset(df):
    out = df.sort_values(["window_id","timestamp"]).copy()
    out["h_offset"] = out.groupby("window_id").cumcount()
    return out

targets = add_hoffset(test_df.query("role=='target'")[["row_id","window_id","building_id","timestamp"]].copy())
targets["building_id"] = targets["building_id"].astype(str)
targets["h_offset"] = targets["h_offset"].astype(int)

sub_hod = pd.read_csv(f"{DATA_DIR}/submission_hod.csv").rename(columns={"meter_reading":"y_hod"})
sub_ttm = pd.read_csv(f"{DATA_DIR}/submission_ttm.csv").rename(columns={"meter_reading":"y_ttm"})
sub_lgb = pd.read_csv(f"{DATA_DIR}/submission_lgbm.csv").rename(columns={"meter_reading_lgbm":"y_lgbm"})
alpha_table = pd.read_parquet(f"{DATA_DIR}/alpha_table_cf.parquet")
bias_table  = pd.read_parquet(f"{DATA_DIR}/bias_table_cf.parquet")

S = (targets.merge(sub_hod, on="row_id", how="left")
            .merge(sub_ttm, on="row_id", how="left")
            .merge(sub_lgb, on="row_id", how="left")
            .merge(alpha_table, on=["building_id","h_offset"], how="left")
            .merge(bias_table,  on=["building_id","h_offset"], how="left"))

S["alpha"] = S["alpha"].fillna(0.0)
S["bias"]  = S["bias"].fillna(0.0)
S["y_hod"]  = pd.to_numeric(S["y_hod"], errors="coerce").fillna(0)
S["y_ttm"]  = pd.to_numeric(S["y_ttm"], errors="coerce").fillna(S["y_hod"])
S["y_lgbm"] = pd.to_numeric(S["y_lgbm"], errors="coerce").fillna(S["y_hod"])

# Base + the exact same 20% LGBM mixing used on TRAIN
W_LGBM = 0.20
y_base = (S["alpha"]*S["y_ttm"] + (1.0-S["alpha"])*S["y_hod"] + S["bias"]).clip(lower=0)
y_pred = np.clip((1.0-W_LGBM)*y_base + W_LGBM*S["y_lgbm"], 0, None)

# Apply isotonic per (building,h) with shrinkage fallbacks
def apply_iso_series(df, y):
    y = y.to_numpy().astype(float)
    out = np.empty_like(y)
    bids = df["building_id"].astype(str).to_numpy()
    hoffs= df["h_offset"].astype(int).to_numpy()
    for i in range(len(y)):
        key = (bids[i], hoffs[i])
        if key in calib_bh:
            ir, rho = calib_bh[key]
        elif bids[i] in calib_b:
            ir, rho = calib_b[bids[i]]
        else:
            ir, rho = iso_global, 0.15  # tiny default shrink
        yhat = float(ir.predict([y[i]])[0])
        out[i] = (1.0 - rho) * y[i] + rho * yhat
    return pd.Series(out, index=df.index)

y_cal = apply_iso_series(S, pd.Series(y_pred))

# Gentle horizon smoothing per window
kernel = np.array([0.2,0.6,0.2])
y_sm = y_cal.to_numpy()
for _, idx in S.groupby("window_id").indices.items():
    v = y_cal.iloc[idx].to_numpy()
    if len(v)>=3:
        y_sm[idx] = np.convolve(v, kernel, mode="same")
y_sm = np.clip(y_sm, 0, None)

# Quantile bands with fallbacks
q_bh = q_bh.copy(); q_b = q_b.copy()  # from previous cell
S2 = (S.merge(q_bh, on=["building_id","h_offset"], how="left")
        .merge(q_b, on="building_id", how="left"))
qlo = S2["q01"].fillna(S2["q01_b"]).fillna(qg01)
qhi = S2["q99"].fillna(S2["q99_b"]).fillna(qg99)

# Main (tight) and safe (looser) clips
y_main  = np.clip(y_sm, 0.98*qlo.to_numpy(), 1.02*qhi.to_numpy())
y_safe  = np.clip(y_sm, 0.95*qlo.to_numpy(), 1.05*qhi.to_numpy())

# No-NaN guard
sub_main = pd.DataFrame({"row_id": S["row_id"], "meter_reading": pd.Series(y_main).fillna(0).clip(lower=0)})
sub_safe = pd.DataFrame({"row_id": S["row_id"], "meter_reading": pd.Series(y_safe).fillna(0).clip(lower=0)})

OUT1 = f"{DATA_DIR}/submission_iso_stack.csv"
OUT2 = f"{DATA_DIR}/submission_iso_stack_loose.csv"
sub_main.sort_values("row_id").to_csv(OUT1, index=False)
sub_safe.sort_values("row_id").to_csv(OUT2, index=False)
print("Wrote:", OUT1, "and", OUT2, "| rows:", len(sub_main))
```

```python
import pandas as pd, numpy as np

DATA_DIR = "/content/drive/My Drive/IISC_data"
K = 15           # neighbors per building
TEMP = 0.15      # softmax temperature (lower = peakier)
EPS = 1e-8

train_df = pd.read_csv(f"{DATA_DIR}/train.csv", parse_dates=["timestamp"])
test_df  = pd.read_csv(f"{DATA_DIR}/test.csv",  parse_dates=["timestamp"])

# Helper: pack last-168h inputs into a fixed-length vector
def pack_inputs(df):
    g = (df.sort_values(["window_id","timestamp"])
           .groupby(["building_id","window_id"])["meter_reading"]
           .apply(lambda s: s.tail(168).to_numpy(np.float32)))
    # keep only vectors of length 168
    g = g[g.map(len)==168]
    X = np.vstack(g.values)
    keys = list(g.index)
    return keys, X

# Helper: pack 24h targets in time order + row_id map for test
def pack_targets_train(df):
    t = (df.sort_values(["window_id","timestamp"])
           .groupby(["building_id","window_id"])["meter_reading"]
           .apply(lambda s: s.to_numpy(np.float32)))
    t = t[t.map(len)==24]
    Y = np.vstack(t.values)
    keys = list(t.index)
    return keys, Y

def pack_targets_test(df):
    t = (df.sort_values(["window_id","timestamp"])
           .groupby(["building_id","window_id"])
           .apply(lambda g: (g["row_id"].to_numpy(np.int64),
                             g["timestamp"].to_numpy())))
    keys, rows = [], []
    for k, (row_ids, ts) in t.items():
        if len(row_ids)==24:
            keys.append(k); rows.append(row_ids)
    R = np.vstack(rows)
    return keys, R

# Build per-building dictionaries
inp_tr = train_df.query("role=='input'")
tgt_tr = train_df.query("role=='target'")
inp_te = test_df.query("role=='input'")
tgt_te = test_df.query("role=='target'")

keys_Xtr, Xtr_all = pack_inputs(inp_tr)
keys_Ytr, Ytr_all = pack_targets_train(tgt_tr)
keys_Xte, Xte_all = pack_inputs(inp_te)
keys_Rte, Rte_all = pack_targets_test(tgt_te)

# Map keys -> rows
from collections import defaultdict
b_tr_X, b_tr_Y, b_te_X, b_te_R = defaultdict(list), defaultdict(list), defaultdict(list), defaultdict(list)

for (b,w), x in zip(keys_Xtr, Xtr_all):
    b_tr_X[str(b)].append((w, x))
for (b,w), y in zip(keys_Ytr, Ytr_all):
    b_tr_Y[str(b)].append((w, y))
for (b,w), x in zip(keys_Xte, Xte_all):
    b_te_X[str(b)].append((w, x))
for (b,w), r in zip(keys_Rte, Rte_all):
    b_te_R[str(b)].append((w, r))

# Make per-building aligned matrices
def to_mat(pairs):
    ws = [w for w,_ in pairs]
    M  = np.vstack([v for _,v in pairs]).astype(np.float32)
    return ws, M

pred_rows, pred_vals = [], []

for b in sorted(b_te_X.keys()):
    if b not in b_tr_X or b not in b_tr_Y:  # no training for this building
        continue

    ws_trX, Xtr = to_mat(b_tr_X[b])
    ws_trY, Ytr = to_mat(b_tr_Y[b])
    ws_teX, Xte = to_mat(b_te_X[b])
    ws_teR, Rte = to_mat(b_te_R[b])

    # align Xtr and Ytr by window_id
    idx = {w:i for i,w in enumerate(ws_trX)}
    keep = [(j, idx[w]) for j,w in enumerate(ws_trY) if w in idx]
    if not keep:
        continue
    jY, iX = zip(*keep)
    Xtr = Xtr[list(iX)]
    Ytr = Ytr[list(jY)]

    # normalize to unit L2 for cosine similarity
    Xtr_n = Xtr / (np.linalg.norm(Xtr, axis=1, keepdims=True) + EPS)
    Xte_n = Xte / (np.linalg.norm(Xte, axis=1, keepdims=True) + EPS)

    # similarity per building via matrix-mul
    S = Xte_n @ Xtr_n.T   # [n_te, n_tr]
    n_te = S.shape[0]

    for j in range(n_te):
        s = S[j]
        # top-K neighbors
        k = min(K, s.size)
        idx_top = np.argpartition(s, -k)[-k:]
        sims = s[idx_top]
        # softmax weights
        w = np.exp(sims / TEMP); w = w / (w.sum() + EPS)
        yhat = (w[:,None] * Ytr[idx_top]).sum(axis=0)  # 24-length
        # attach row_ids in timestamp order
        row_ids = Rte[j]
        pred_rows.append(row_ids)
        pred_vals.append(np.clip(yhat, 0, None))

# Assemble submission from kNN
row_ids_all = np.concatenate(pred_rows, axis=0)
vals_all    = np.concatenate(pred_vals, axis=0)
sub_knn = pd.DataFrame({"row_id": row_ids_all, "meter_reading": vals_all}).sort_values("row_id").reset_index(drop=True)
knn_path = f"{DATA_DIR}/submission_knn_building.csv"
sub_knn.to_csv(knn_path, index=False)
print("Wrote:", knn_path, "| rows:", len(sub_knn))
```

```python
# === 2) UTILITIES ===
EPS = 1e-8

def robust_standardize(x: np.ndarray) -> np.ndarray:
    """Shape-only standardization: median/MAD (robust to level outliers)."""
    med = np.median(x)
    mad = np.median(np.abs(x - med)) + 1e-6
    return (x - med) / mad

def day_chunks_7(x168: np.ndarray):
    """Split 168h into 7x24; return (day_sums, list_of_24h_arrays)."""
    days = [x168[i*24:(i+1)*24] for i in range(7)]
    return [float(d.sum()) for d in days], days

def sum_preserving_smooth24(y: np.ndarray) -> np.ndarray:
    """3-tap smoothing with mirrored edges, preserving 24h sum."""
    y = y.astype(np.float32)
    s_old = float(y.sum())
    z = y.copy()
    out = y.copy()
    for h in range(24):
        left  = z[h-1] if h>0 else z[1]
        mid   = z[h]
        right = z[h+1] if h<23 else z[22]
        out[h]  = 0.25*left + 0.5*mid + 0.25*right
    s_new = float(out.sum()) + EPS
    return out * (s_old / s_new)
```

```python
# === 3) PACKING HELPERS (robust to timestamp dtype) ===
def pack_inputs_with_ts(df: pd.DataFrame):
    g = (df.sort_values(["window_id","timestamp"])
           .groupby(["building_id","window_id"]))
    X, last_ts, keys = [], [], []
    for (b,w), gi in g:
        ts = pd.to_datetime(gi["timestamp"], errors="coerce")
        v = gi["meter_reading"].tail(168).to_numpy(np.float32)
        if len(v)==168:
            X.append(v)
            last_ts.append(ts.max())              # last observed hour
            keys.append((str(b), int(w)))
    return keys, np.vstack(X), np.array(last_ts, dtype="datetime64[ns]")

def pack_targets_train_with_ts(df: pd.DataFrame):
    g = (df.sort_values(["window_id","timestamp"])
           .groupby(["building_id","window_id"]))
    Y, TS, keys = [], [], []
    for (b,w), gi in g:
        ts = pd.to_datetime(gi["timestamp"], errors="coerce").to_numpy()
        v  = gi["meter_reading"].to_numpy(np.float32)
        if len(v)==24:
            Y.append(v); TS.append(ts); keys.append((str(b), int(w)))
    return keys, np.vstack(Y), TS

def pack_targets_test_with_ts(df: pd.DataFrame):
    g = (df.sort_values(["window_id","timestamp"])
           .groupby(["building_id","window_id"]))
    R, TS, keys = [], [], []
    for (b,w), gi in g:
        ts = pd.to_datetime(gi["timestamp"], errors="coerce").to_numpy()
        rid = gi["row_id"].to_numpy(np.int64)
        if len(rid)==24:
            keys.append((str(b), int(w))); R.append(rid); TS.append(ts)
    return keys, np.vstack(R), TS
```

```python
# === 4) BUILD SPLITS & ALIGN ===
inp_tr = train_df.query("role=='input'")
tgt_tr = train_df.query("role=='target'")
inp_te = test_df.query("role=='input'")
tgt_te = test_df.query("role=='target'")

keys_Xtr, Xtr_all, Xtr_last_ts = pack_inputs_with_ts(inp_tr)
keys_Ytr, Ytr_all, Ytr_TS      = pack_targets_train_with_ts(tgt_tr)
keys_Xte, Xte_all, Xte_last_ts = pack_inputs_with_ts(inp_te)
keys_Rte, Rte_all, Rte_TS      = pack_targets_test_with_ts(tgt_te)

# Align Xtr (168) with Ytr (24) by (building, window)
idx_Xtr = {k:i for i,k in enumerate(keys_Xtr)}
keep = [(j, idx_Xtr[k]) for j,k in enumerate(keys_Ytr) if k in idx_Xtr]
if not keep:
    raise RuntimeError("No overlapping (building,window) between Xtr and Ytr")

jY, iX = zip(*keep)
Xtr = Xtr_all[list(iX)]
Ytr = Ytr_all[list(jY)]
Ytr_ts = [Ytr_TS[j] for j in jY]
Ytr_b  = [keys_Ytr[j][0] for j in jY]  # building ids aligned

print("Aligned train windows:", Xtr.shape, Ytr.shape)
```

```python
# === 5) BIAS TABLE (bld × HOD) + SAFE CAPS ===
from collections import defaultdict

# Weekly-copy pseudo-forecast: first 24h from last-168
WC  = np.stack([x[:24] for x in Xtr], axis=0)   # 24
RES = Ytr - WC                                  # residuals

bias_table = defaultdict(dict)   # bias_table[bld][hour] = mean residual
cap_table  = {}                  # cap per building (p99.7 of true)

for b in np.unique(Ytr_b):
    mask = np.array([bb==b for bb in Ytr_b], dtype=bool)
    yb   = Ytr[mask]          # [n_win_b, 24]
    wb   = RES[mask]          # [n_win_b, 24]
    tsb  = np.array(Ytr_ts, dtype=object)[mask]  # list of arrays (timestamps)

    # hour-of-day bias, robust to dtype
    hod_vals, res_vals = [], []
    for win_idx in range(wb.shape[0]):
        ts_vec = pd.to_datetime(tsb[win_idx], errors="coerce")
        valid = ~pd.isna(ts_vec)
        if valid.any():
            hod = ts_vec[valid].hour.to_numpy(dtype=np.int16)
            res = wb[win_idx][valid]
            hod_vals.append(hod)
            res_vals.append(res)
    if len(hod_vals):
        hod_vals = np.concatenate(hod_vals)
        res_vals = np.concatenate(res_vals)
    else:
        hod_vals = np.array([], dtype=np.int16)
        res_vals = np.array([], dtype=np.float32)

    for h in range(24):
        m = (hod_vals==h)
        bias_table[b][h] = float(res_vals[m].mean()) if m.any() else 0.0

    # building cap
    cap_table[b] = float(np.quantile(yb.flatten(), 0.997))

print("Bias/cap tables built for buildings:", len(cap_table))
```

```python
# === 6) IMPROVED KNN (shape sim + weekend parity + energy recalib + sim floor) ===
from collections import defaultdict

# Group per building for fast access
b_tr_X  = defaultdict(list)  # (w, x168, last_ts)
b_tr_YT = defaultdict(list)  # (w, y24, first_target_ts, is_weekend)
b_te_XR = defaultdict(list)  # (w, x168, row_ids24, first_target_ts, is_weekend)

for (b,w), x, ts_last in zip(keys_Xtr, Xtr_all, Xtr_last_ts):
    b_tr_X[str(b)].append((w, x, ts_last))

for (b,w), y, ts in zip(keys_Ytr, Ytr_all, Ytr_TS):
    ts0 = pd.to_datetime(ts, errors="coerce")[0]
    weekend = (ts0.weekday()>=5) if pd.notna(ts0) else False
    b_tr_YT[str(b)].append((w, y, ts0, weekend))

for (b,w), x, rid, ts in zip(keys_Xte, Xte_all, Rte_all, Rte_TS):
    ts_arr = pd.to_datetime(ts, errors="coerce")
    ts0 = ts_arr[0] if len(ts_arr) else pd.NaT
    weekend = (ts0.weekday()>=5) if pd.notna(ts0) else False
    b_te_XR[str(b)].append((w, x, rid, ts0, weekend))

# KNN params (safe defaults)
K_GLOBAL = 9
TEMP_GLOBAL = 0.12
SIM_FLOOR = 0.50

pred_rows, pred_vals = [], []

for b in sorted(b_te_XR.keys()):
    if b not in b_tr_X or b not in b_tr_YT:
        continue

    # Build aligned train matrices for this building
    trX_pairs = {w:(x,ts_last) for (w,x,ts_last) in b_tr_X[b]}
    trY_list  = []
    for (w,y,ts0,weekend) in b_tr_YT[b]:
        if w in trX_pairs:
            x, ts_last = trX_pairs[w]
            trY_list.append((w, x, y, ts0, weekend))
    if not trY_list:
        continue

    ws_tr = [w for (w,_,_,_,_) in trY_list]
    Xtr_b = np.vstack([x for (_,x,_,_,_) in trY_list]).astype(np.float32)
    Ytr_b = np.vstack([y for (_,_,y,_,_) in trY_list]).astype(np.float32)
    Ytr_weekend = np.array([wk for (_,_,_,_,wk) in trY_list], dtype=bool)

    # Shape-only standardization for similarity
    Xtr_std = np.vstack([robust_standardize(x) for x in Xtr_b])
    Xtr_std = Xtr_std / (np.linalg.norm(Xtr_std, axis=1, keepdims=True) + EPS)

    for (wte, xte, rid24, ts0, is_weekend) in b_te_XR[b]:
        # std + cosine sim
        xte_std = robust_standardize(xte)
        xte_std = xte_std / (np.linalg.norm(xte_std) + EPS)

        # weekend parity filter (ensure minimum pool size)
        mask = (Ytr_weekend == is_weekend)
        if mask.sum() < 200:  # relax if thin
            mask = np.ones_like(Ytr_weekend, dtype=bool)

        S = (Xtr_std[mask] @ xte_std)           # cosine similarities
        k = min(K_GLOBAL, S.size)
        if k == 0:
            # no neighbors; fallback to weekly copy
            yhat = xte[:24].astype(np.float32)
        else:
            idx_top = np.argpartition(S, -k)[-k:]
            sims = S[idx_top]
            s_max = float(sims.max()) if sims.size else 0.0

            if s_max < SIM_FLOOR:
                yhat = xte[:24].astype(np.float32)
            else:
                wts = np.exp(sims / TEMP_GLOBAL)
                wts = wts / (wts.sum() + EPS)
                yhat = (wts[:,None] * Ytr_b[mask][idx_top]).sum(axis=0).astype(np.float32)

        # Day-sum recalibration (match expected energy safely)
        dsums, days = day_chunks_7(xte)
        target_sum = float(dsums[0]) if len(dsums)==7 else float(np.median(dsums))
        scale = np.clip(target_sum / (float(yhat.sum()) + EPS), 0.7, 1.3)
        yhat = np.clip(yhat * scale, 0, None)

        pred_rows.append(rid24)
        pred_vals.append(yhat)

sub_knn = pd.DataFrame({
    "row_id": np.concatenate(pred_rows),
    "meter_reading": np.concatenate(pred_vals).astype(np.float32)
}).sort_values("row_id").reset_index(drop=True)

knn_path = f"{DATA_DIR}/submission_knn_improved.csv"
sub_knn.to_csv(knn_path, index=False)
print("Wrote:", knn_path, "| rows:", len(sub_knn))
```

```python
# === 7) LOAD BASE SUBMISSIONS & HOUR-GATED BLEND ===
med = pd.read_csv(f"{DATA_DIR}/submission_ens_median_top.csv").sort_values("row_id").rename(columns={"meter_reading":"m_med"})
lgb = pd.read_csv(f"{DATA_DIR}/submission_lgbm.csv").sort_values("row_id")
lgb = lgb.rename(columns={"meter_reading_lgbm":"m_lgbm"}) if "meter_reading_lgbm" in lgb.columns else lgb.rename(columns={"meter_reading":"m_lgbm"})
knn = pd.read_csv(f"{DATA_DIR}/submission_knn_improved.csv").sort_values("row_id").rename(columns={"meter_reading":"m_knn"})

base = med.merge(lgb[["row_id","m_lgbm"]], on="row_id", how="left") \
          .merge(knn[["row_id","m_knn"]],  on="row_id", how="left")

# Map row_id -> (building_id, hod) from test targets
meta_te = (tgt_te[["row_id","building_id","timestamp"]]
           .dropna()
           .sort_values("row_id"))
meta_te["hod"] = pd.to_datetime(meta_te["timestamp"], errors="coerce").dt.hour

dfb = base.merge(meta_te[["row_id","building_id","hod"]], on="row_id", how="left")
dfb["m_lgbm"] = pd.to_numeric(dfb["m_lgbm"], errors="coerce").fillna(dfb["m_med"])
dfb["m_knn"]  = pd.to_numeric(dfb["m_knn"],  errors="coerce").fillna(dfb["m_med"])

def weights_for_hod(h: int):
    # heuristic gates: adjust if your CV shows different regimes
    if 0 <= h <= 5:      # night
        return 0.55, 0.35, 0.10   # MED, LGBM, KNN
    elif 6 <= h <= 11:   # morning ramp
        return 0.35, 0.55, 0.10
    elif 12 <= h <= 17:  # afternoon
        return 0.60, 0.30, 0.10
    else:                # evening ramp
        return 0.35, 0.55, 0.10

hod = dfb["hod"].fillna(12).astype(int).to_numpy()
wM = np.empty_like(hod, dtype=np.float32)
wL = np.empty_like(hod, dtype=np.float32)
wK = np.empty_like(hod, dtype=np.float32)
for i, h in enumerate(hod):
    m,l,k = weights_for_hod(h)
    wM[i], wL[i], wK[i] = m,l,k

y_blend = (wM*dfb["m_med"].values + wL*dfb["m_lgbm"].values + wK*dfb["m_knn"].values).astype(np.float32)
dfb["y_blend"] = np.clip(y_blend, 0, None)
```

```python
# === 8) POST-PROCESS PER (BUILDING, WINDOW): BIAS -> SMOOTH -> CAP -> CLIP ===
rows_pp, vals_pp = [], []

g = (tgt_te.sort_values(["building_id","window_id","timestamp"])
            .groupby(["building_id","window_id"]))

for (b,w), gi in g:
    b = str(b)
    rows  = gi["row_id"].to_numpy()
    ts24  = pd.to_datetime(gi["timestamp"], errors="coerce")
    hod24 = ts24.dt.hour.to_numpy(dtype=np.int16)

    y     = dfb.loc[dfb["row_id"].isin(rows), "y_blend"].to_numpy(dtype=np.float32)

    # 1) subtract train-time bias[b][hod]
    if b in bias_table:
        bias_vec = np.array([bias_table[b].get(int(h), 0.0) for h in hod24], dtype=np.float32)
        y = y - bias_vec

    # 2) smooth with sum preservation
    y = sum_preserving_smooth24(y)

    # 3) cap by building safe p99.7 (from train)
    cap = cap_table.get(b, np.float32(np.inf))
    y = np.minimum(y, cap)

    # 4) non-negativity
    y = np.clip(y, 0, None)

    rows_pp.append(rows)
    vals_pp.append(y)

pred_pp = pd.DataFrame({
    "row_id": np.concatenate(rows_pp),
    "meter_reading": np.concatenate(vals_pp).astype(np.float32)
}).sort_values("row_id").reset_index(drop=True)

final_path = f"{DATA_DIR}/submission_blend_biassmoothcap.csv"
pred_pp.to_csv(final_path, index=False)
print("Wrote:", final_path, "| rows:", len(pred_pp))
```

```python
# Cache last-168 per (bld,win) from test inputs; also compute rampiness quantile
g = (inp_te.sort_values(["building_id","window_id","timestamp"])
          .groupby(["building_id","window_id"])["meter_reading"])

x168_map = {}
ramp_scores = []

for (b,w), s in g:
    v = s.tail(168).to_numpy(np.float32)
    if len(v) == 168:
        x168_map[(str(b), int(w))] = v
        # rampiness: max abs diff of last day (recent 24h)
        last24 = v[-24:]
        ramp_scores.append(float(np.max(np.abs(np.diff(last24)))))

# thresholds to detect "rampy" days
q75 = np.quantile(ramp_scores, 0.75) if ramp_scores else 0.0
q90 = np.quantile(ramp_scores, 0.90) if ramp_scores else 0.0
print("Ramp thresholds:", q75, q90, "| cached windows:", len(x168_map))
```

```python
def robust_standardize(x):
    med = np.median(x)
    mad = np.median(np.abs(x - med)) + 1e-6
    return (x - med) / mad

def cos_sim(a, b):
    na = np.linalg.norm(a); nb = np.linalg.norm(b)
    if na < 1e-8 or nb < 1e-8: return 0.0
    return float(np.dot(a, b) / (na * nb))

def day_chunks_7(x168):
    return [x168[i*24:(i+1)*24] for i in range(7)]

def adaptive_blend_window(y_base_24, x168, ramp_q75, ramp_q90):
    """
    y_base_24: base 24-vector (MED⊕LGBM w=0.235)
    x168: last-168 vector
    Returns: adjusted 24-vector (no smoothing, clipped >=0)
    """
    days = day_chunks_7(x168)    # [d0 ... d6], d6 is most recent day
    if len(days) != 7 or any(len(d)!=24 for d in days):
        return np.clip(y_base_24, 0, None)

    prev_day = days[-1]
    weekago  = days[0]

    # 1) ramp-aware LGBM weight tweak around your 0.235
    ramp_score = float(np.max(np.abs(np.diff(prev_day))))
    if ramp_score >= ramp_q90:
        w_lgbm = 0.32   # very rampy → trust LGBM more
    elif ramp_score >= ramp_q75:
        w_lgbm = 0.27
    else:
        w_lgbm = 0.20   # flat → lean safer

    # reconstruct an *adaptive* base from the two columns we have
    # NOTE: we don't have per-hour separate med/lgbm anymore here,
    # so we start from provided y_base_24 (already at 0.235)
    # and nudge toward or away from LGBM by mixing with weekly copy.
    y = y_base_24.copy()

    # 2) tiny weekly-copy assist only if shapes align
    c = cos_sim(robust_standardize(prev_day), robust_standardize(weekago))
    energy_ratio = (weekago.sum()+1e-6)/(prev_day.sum()+1e-6)
    if c >= 0.88 and 0.85 <= energy_ratio <= 1.15:
        alpha_wc = min(0.12, max(0.0, (c - 0.88) / 0.12))  # 0..0.12 as c goes 0.88→1.0
        y = (1.0 - alpha_wc)*y + alpha_wc*weekago

    # 3) energy calibration to robust next-day expectation:
    # use median of last 3 day sums (prev 1–3 days), safer than week-ago
    day_sums = np.array([d.sum() for d in days[-3:]], dtype=np.float32)
    target_sum = float(np.median(day_sums))
    scale = np.clip(target_sum / (float(y.sum()) + 1e-8), 0.90, 1.10)
    y = np.clip(y * scale, 0, None)

    return y.astype(np.float32)
```

```python
def robust_standardize(x):
    med = np.median(x); mad = np.median(np.abs(x - med)) + 1e-6
    return (x - med) / mad

def cos_sim(a, b):
    na = np.linalg.norm(a); nb = np.linalg.norm(b)
    if na < 1e-8 or nb < 1e-8: return 0.0
    return float(np.dot(a, b) / (na * nb))

def day_chunks_7(x168):  # [24]*7
    return [x168[i*24:(i+1)*24] for i in range(7)]

def anti_spike_isolated(y, factor=1.80, max_fixes=2):
    """Trim at most 2 isolated needles; no sum re-match."""
    y = y.astype(np.float32).copy()
    fixes = 0
    for i in range(1, 23):
        if fixes >= max_fixes: break
        if y[i] > max(y[i-1], y[i+1]) * factor:
            y[i] = 0.5 * (y[i-1] + y[i+1])
            fixes += 1
    return y

def window_adjust(y_med24, y_lgb24, x168, W, alpha_max, c_thresh, e_tol, ramp_q75, ramp_q90,
                  use_knn=False, y_knn24=None, knn_w=0.05, knn_c=0.93, knn_sum_lo=0.98, knn_sum_hi=1.02,
                  use_antispike=False):
    """
    Build per-window 24h prediction with tiny, safe adjustments.
    y_med24, y_lgb24: arrays length 24 (from MED/LGBM for this window)
    x168: last-168 vector
    W: global LGBM weight
    alpha_max: max weekly-copy assist (<=0.12)
    c_thresh: cosine threshold for allowing weekly assist (0.90–0.92+)
    e_tol: energy cal tolerance, e.g., 0.08 → ±8%
    """
    y = np.clip((1.0 - W)*y_med24 + W*y_lgb24, 0, None).astype(np.float32)

    days = day_chunks_7(x168)
    if len(days)==7 and all(len(d)==24 for d in days):
        prev_day, weekago = days[-1], days[0]

        # ramp-aware nudge of W around given global W (±0.03 band)
        ramp = float(np.max(np.abs(np.diff(prev_day))))
        if ramp >= ramp_q90:   W_eff = min(W+0.03, 0.35)
        elif ramp >= ramp_q75: W_eff = min(W+0.02, 0.35)
        else:                  W_eff = max(W-0.02, 0.15)
        y = np.clip((1.0 - W_eff)*y_med24 + W_eff*y_lgb24, 0, None)

        # weekly-copy assist (shape+level agreement only)
        c = cos_sim(robust_standardize(prev_day), robust_standardize(weekago))
        er = (weekago.sum()+1e-6)/(prev_day.sum()+1e-6)
        if c >= c_thresh and (1.0 - e_tol) <= er <= (1.0 + e_tol):
            # scale alpha smoothly from threshold to 1.0
            alpha = np.clip((c - c_thresh) / (1.0 - c_thresh + 1e-9), 0.0, 1.0) * alpha_max
            y = (1.0 - alpha)*y + alpha*weekago

        # soft per-window energy calibration to robust expectation (median of last 3 days)
        target_sum = float(np.median([d.sum() for d in days[-3:]]))
        scale = np.clip(target_sum / (float(y.sum()) + 1e-8), 1.0 - e_tol, 1.0 + e_tol)
        y = y * scale

    # optional guarded KNN
    if use_knn and y_knn24 is not None and len(y_knn24)==24:
        c2 = cos_sim(y, y_knn24)
        sr = float((y_knn24.sum()+1e-6)/(y.sum()+1e-6))
        if c2 >= knn_c and knn_sum_lo <= sr <= knn_sum_hi:
            y = (1.0 - knn_w)*y + knn_w*y_knn24

    # optional anti-spike on isolated needles
    if use_antispike:
        y = anti_spike_isolated(y, factor=1.80, max_fixes=2)

    return np.clip(y, 0, None).astype(np.float32)
```

```python
import numpy as np
import pandas as pd

# =========================
# Config / Paths
# =========================
DATA_DIR = "/content/drive/My Drive/IISC_data"  # <-- change if needed
TRAIN_PATH = f"{DATA_DIR}/train.csv"
TEST_PATH  = f"{DATA_DIR}/test.csv"
SUB_MED    = f"{DATA_DIR}/submission_ens_median_top.csv"
SUB_LGBM   = f"{DATA_DIR}/submission_lgbm.csv"
OUT_PATH   = f"{DATA_DIR}/submission_nnls_hour_weights.csv"

# =========================
# 1) Load data
# =========================
train = pd.read_csv(TRAIN_PATH, parse_dates=["timestamp"])
test  = pd.read_csv(TEST_PATH,  parse_dates=["timestamp"])

# Two base TEST submissions (already computed by you)
sub_med = (pd.read_csv(SUB_MED)
             .sort_values("row_id")
             .rename(columns={"meter_reading":"m_med"}))
sub_lgb = pd.read_csv(SUB_LGBM).sort_values("row_id")
sub_lgb = (sub_lgb.rename(columns={"meter_reading_lgbm":"m_lgbm"})
           if "meter_reading_lgbm" in sub_lgb.columns
           else sub_lgb.rename(columns={"meter_reading":"m_lgbm"}))

# =========================
# 2) Build train rolling windows (168 -> 24) without .apply()
# =========================
def pack_train_windows(df: pd.DataFrame):
    """Return aligned train windows:
       keys: list[(building_id, window_id)]
       X:    (n_win, 168) last-168 inputs
       Y:    (n_win, 24)  next-24 true targets
       T0:   list of first target timestamps per window (for reference)
    """
    df = df.sort_values(["building_id","window_id","timestamp"])
    inp = df[df["role"] == "input"]
    tgt = df[df["role"] == "target"]

    # last 168 hours per (bld, win)
    X_dict = {}
    for (b,w), gi in inp.groupby(["building_id","window_id"], sort=False):
        v = gi["meter_reading"].to_numpy(np.float32)
        if len(v) >= 168:
            X_dict[(str(b), int(w))] = v[-168:]

    # next 24 hours per (bld, win)
    Y_dict, T0_dict = {}, {}
    for (b,w), gi in tgt.groupby(["building_id","window_id"], sort=False):
        v = gi["meter_reading"].to_numpy(np.float32)
        if len(v) == 24:
            key = (str(b), int(w))
            Y_dict[key]  = v
            T0_dict[key] = pd.to_datetime(gi["timestamp"], errors="coerce").iloc[0]

    # align
    keys = sorted(set(X_dict.keys()) & set(Y_dict.keys()))
    X = np.stack([X_dict[k] for k in keys], axis=0)
    Y = np.stack([Y_dict[k] for k in keys], axis=0)
    T0 = [T0_dict[k] for k in keys]
    return keys, X, Y, T0

keys_tr, Xtr, Ytr, T0 = pack_train_windows(train)
print("Aligned train windows:", Xtr.shape, Ytr.shape)

# =========================
# 3) Proxies on train (leak-safe, simple)
# =========================
def weekly_copy_24(x168: np.ndarray) -> np.ndarray:
    """MED proxy: first 24h of the last-168 (i.e., same hours a week ago)."""
    return x168[:24].astype(np.float32)

def lgbm_proxy_from_x(x168: np.ndarray) -> np.ndarray:
    """LGBM proxy: tiny fixed linear combo of lags {1, 24, 168}.
       This approximates a lags-ridge and tracks LGBM's behavior reasonably well.
    """
    last24 = x168[-24:]
    lag1   = np.r_[last24[-1], last24[:-1]]
    lag24  = x168[-48:-24]
    lag168 = x168[:24]
    return (0.50*lag1 + 0.30*lag24 + 0.20*lag168).astype(np.float32)

med_proxy = np.stack([weekly_copy_24(x)   for x in Xtr], axis=0)
lgb_proxy = np.stack([lgbm_proxy_from_x(x) for x in Xtr], axis=0)

# =========================
# 4) Learn hour-of-day weights (NNLS + sum-to-one via tiny grid)
# =========================
# For each hour h, solve:
#   min_{w in [0,1]}  MSE( y_h , (1-w)*med_h + w*lgb_h )
# This is equivalent to non-negative weights that sum to 1.
W_h = np.zeros(24, dtype=np.float32)
grid = np.linspace(0.00, 1.00, 101)  # 0..1 in 0.01 steps

for h in range(24):
    y = Ytr[:, h]
    m = med_proxy[:, h]
    l = lgb_proxy[:, h]
    best_w, best_mse = 0.5, 1e18
    # try all convex weights on a fine grid
    for w in grid:
        yhat = (1.0 - w)*m + w*l
        mse  = np.mean((y - yhat)**2)
        if mse < best_mse:
            best_mse, best_w = mse, float(w)
    W_h[h] = best_w

print("Learned LGBM hour-weights:", np.round(W_h, 3))

# =========================
# 5) Apply learned hour-weights to TEST submissions
# =========================
test_tgt = test[test["role"]=="target"][["row_id","timestamp"]].copy()
test_tgt["hod"] = pd.to_datetime(test_tgt["timestamp"], errors="coerce").dt.hour.astype(int)

sub = (sub_med.merge(sub_lgb[["row_id","m_lgbm"]], on="row_id", how="left")
              .merge(test_tgt[["row_id","hod"]],       on="row_id", how="left")
              .sort_values("row_id"))
sub["m_lgbm"] = pd.to_numeric(sub["m_lgbm"], errors="coerce").fillna(sub["m_med"]).astype(np.float32)
sub["m_med"]  = pd.to_numeric(sub["m_med"],  errors="coerce").astype(np.float32)

# map hod -> weight, then blend
w_map = {h: W_h[h] for h in range(24)}
w_vec = sub["hod"].map(w_map).astype(np.float32).to_numpy()

y_pred = (1.0 - w_vec) * sub["m_med"].to_numpy() + w_vec * sub["m_lgbm"].to_numpy()
y_pred = np.clip(y_pred.astype(np.float32), 0, None)

out = pd.DataFrame({"row_id": sub["row_id"], "meter_reading": y_pred}).sort_values("row_id").reset_index(drop=True)
out.to_csv(OUT_PATH, index=False)
print(f"Wrote: {OUT_PATH} | rows: {len(out)} | min/max: {float(out.meter_reading.min()):.6f}/{float(out.meter_reading.max()):.6f}")
```

```python
import numpy as np, pandas as pd
from numpy.linalg import lstsq

DATA_DIR = "/content/drive/My Drive/IISC_data"  # adjust

# --- Load base submissions (unchanged) ---
med = pd.read_csv(f"{DATA_DIR}/submission_ens_median_top.csv").sort_values("row_id").rename(columns={"meter_reading":"m_med"})
lgb = pd.read_csv(f"{DATA_DIR}/submission_lgbm.csv").sort_values("row_id")
lgb = lgb.rename(columns={"meter_reading_lgbm":"m_lgbm"}) if "meter_reading_lgbm" in lgb.columns else lgb.rename(columns={"meter_reading":"m_lgbm"})

df_base = med.merge(lgb[["row_id","m_lgbm"]], on="row_id", how="left")
df_base["m_med"]  = pd.to_numeric(df_base["m_med"],  errors="coerce").astype(np.float32)
df_base["m_lgbm"] = pd.to_numeric(df_base["m_lgbm"], errors="coerce").fillna(df_base["m_med"]).astype(np.float32)

# Exact best baseline (your winner): w=0.235
W = 0.235
df_base["y_base"] = np.clip((1.0 - W)*df_base["m_med"].to_numpy() + W*df_base["m_lgbm"].to_numpy(), 0, None).astype(np.float32)

# --- Load test for grouping and last-168 inputs ---
test = pd.read_csv(f"{DATA_DIR}/test.csv", parse_dates=["timestamp"])
inp_te = test.query("role=='input'").copy()
tgt_te = test.query("role=='target'").copy()

# cache last-168 per (building, window)
x168_map = {}
for (b,w), gi in inp_te.sort_values(["building_id","window_id","timestamp"]).groupby(["building_id","window_id"], sort=False):
    v = gi["meter_reading"].tail(168).to_numpy(np.float32)
    if len(v)==168:
        x168_map[(str(b), int(w))] = v

# --- Fourier+trend extrapolation from last-168 to next-24 ---
def fourier_forecast_24(y168, daily_K=3, weekly_K=1):
    """
    Fit LS on last 168 samples with:
    intercept + linear trend + daily harmonics (K) + weekly harmonics (K2),
    then forecast the next 24 steps.
    """
    n = 168
    t = np.arange(n, dtype=np.float32)

    # design on history
    cols = [np.ones(n, np.float32), t]  # intercept + trend

    # daily (period=24)
    for k in range(1, daily_K+1):
        ang = 2*np.pi*k*t/24.0
        cols.append(np.sin(ang)); cols.append(np.cos(ang))

    # weekly (period=168)
    for k in range(1, weekly_K+1):
        ang = 2*np.pi*k*t/168.0
        cols.append(np.sin(ang)); cols.append(np.cos(ang))

    X = np.vstack(cols).T  # [168, p]
    beta, *_ = lstsq(X, y168, rcond=None)

    # design for forecast horizon
    t_f = np.arange(n, n+24, dtype=np.float32)
    cols_f = [np.ones_like(t_f), t_f]
    for k in range(1, daily_K+1):
        ang = 2*np.pi*k*t_f/24.0
        cols_f.append(np.sin(ang)); cols_f.append(np.cos(ang))
    for k in range(1, weekly_K+1):
        ang = 2*np.pi*k*t_f/168.0
        cols_f.append(np.sin(ang)); cols_f.append(np.cos(ang))
    Xf = np.vstack(cols_f).T  # [24, p]

    yhat = Xf @ beta
    return yhat.astype(np.float32)

def cos_sim(a, b):
    a = a.astype(np.float32); b = b.astype(np.float32)
    na = np.linalg.norm(a); nb = np.linalg.norm(b)
    if na < 1e-8 or nb < 1e-8: return 0.0
    return float(a @ b / (na*nb))

# Build per-window base + guarded Fourier
base = (df_base
        .merge(tgt_te[["row_id","building_id","window_id","timestamp"]], on="row_id", how="left")
        .sort_values(["building_id","window_id","timestamp"]))

rows_out, vals_out = [], []

# guard thresholds (very strict)
CS_MIN   = 0.93   # cosine similarity between base and Fourier shape
SUM_LOW  = 0.97   # day-sum ratio guard
SUM_HIGH = 1.03
W_FOUR   = 0.05   # at most +5% Fourier when it agrees

for (b,w), gi in base.groupby(["building_id","window_id"], sort=False):
    key = (str(b), int(w))
    rows = gi["row_id"].to_numpy()
    yb   = gi["y_base"].to_numpy(np.float32)

    y = yb
    if key in x168_map and len(yb)==24:
        y_four = fourier_forecast_24(x168_map[key], daily_K=3, weekly_K=1)
        # soft non-negativity
        y_four = np.clip(y_four, 0, None)

        # agreement checks
        cs = cos_sim(yb, y_four)
        sr = float((y_four.sum()+1e-6)/(yb.sum()+1e-6))

        if (cs >= CS_MIN) and (SUM_LOW <= sr <= SUM_HIGH):
            y = (1.0 - W_FOUR)*yb + W_FOUR*y_four  # tiny convex tap
        # else keep base

    y = np.clip(y, 0, None)
    rows_out.append(rows); vals_out.append(y.astype(np.float32))

sub = pd.DataFrame({
    "row_id": np.concatenate(rows_out),
    "meter_reading": np.concatenate(vals_out)
}).sort_values("row_id").reset_index(drop=True)

path = f"{DATA_DIR}/submission_base_plus_fourier_guarded.csv"
sub.to_csv(path, index=False)
print("Wrote:", path, "| rows:", len(sub), "| min/max:", float(sub.meter_reading.min()), float(sub.meter_reading.max()))
```
